### Standard math functions

| function | purpose |
| - | - |

| `math.cbrt(`*n*`)` | cube root |

````python
# square root
import math
print(math.sqrt(1764))
````

````python
# cube root
import math
print(math.cbrt(74088))
````

````python
# logarithm, base e
import math
print(math.log(42))
````

````python
# logarithm, base 10
import math
print(math.log(42, 10))
````

````python
# e raised to power x
import math
print(math.exp(3.738))
````

````python
# 10 raised to power x
import math
print(math.pow(10, 1.624))
````

````python
# remove a fraction from a float
import math
print(math.trunc(42.2))
````

### Probabability functions
Theee are three math library functions related to probability:
| method | function | purpose |
| - | - | - |
| Factorial | `factorial(`*n*`)` | the number of ways to order n numbers |
| Permutations | `perm(`*n*`, `*k*`)` | the number of ways to order k items out of n items |
| Combinations | `comb(`*n*`, `*k*`)` | the number of ways to choose k items from n |




### Factorial
Factorial is the number of ways to order n numbers:
$$n!$$

````python
import math
# 5 factorial
print(math.factorial(5))
````

__Exercise__: compute $7!$.

### Permutations
Permutations are the number of ways to order k items from n items:
$$ {^n\mathrm{P}_k} = \frac{n!}{(n-k)!} $$


````python
import math
# permutations of 3 items out of 7
print(math.perm(7, 3))
````

__Exercise__: compute ${^{10}\mathrm{P}_4}$.

### Combinations
Combinations are the number of ways to choose k items from n  
$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

````python
# combinations of choosing k items from n
import math
# combinations of 3 items out of 7
print(math.comb(7, 3))
````

## [Statistics](https://docs.python.org/3/library/statistics.html#statistics) - `statistics` library
Python has a wide variety of standard statistics functions.
To use it, we import the `statistics` module:

```import statistics```

These are four statistics functions useful in math formulas.
| function | purpose |
| - | - |
| 'mean()` | Mean (“average”) |
| 'median()` | Median (middle value) |
| 'mode()` | Single mode (most common value) |
| 'stdev()` | Standard deviation |

````python
# mean
import statistics
data = [15, 21, 28, 33, 40, 44, 49, 56, 62, 72]
print(statistics.mean(data))
````

````python
# median
import statistics
data = [42.0, 250.5, 12.4, 42.0, 98.7, 12.4, 42.0, 86.4, 134.8, 3.7, 111.6]
print(statistics.median(data))
````

````python
# mode
import statistics
data = [42.0, 250.5, 12.4, 42.0, 98.7, 12.4, 42.0, 86.4, 134.8, 3.7, 111.6]
print(statistics.mode(data))
````

````python
# standard deviation
import statistics
data = [42.0, 250.5, 12.4, 42.0, 98.7, 12.4, 42.0, 86.4, 134.8, 3.7, 111.6]
print(statistics.stdev(data))
````
