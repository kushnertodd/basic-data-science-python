# Variables
The most basic program statement assigns a value to a variable with `=`.
A variable initially is empty.
When a variable is assigned, it is given a small place in memory to store a value.
Assigning a variable to a basic data type value, `int`, `float`, `boolean`, or `string`, essentially places the value into that memory location.
That variable is placed into the data state.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-2.jpg"
alt="Assigning variables" width="200" height="auto">



Variables are assigned to the value of *expressions*.
An expression is a combination of variables, constants, operators, and functions that produce a value.
Expressions follow the rules of standard mathematical statements.
Any place a variable or function appears in an expression, its value is substituted.

```python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2, height = 21)
area = width * height
# data state: (width = 2, height = 21, area = 42)
print(area)
```

A variable must be assigned a value and placed into the data state before it is used in an expression, else it gives an error. This commonly occurs when a variable name is misspelled.

```python
# data state: ()
# print(p) -- this gives an error, p must be assigned a value before it is used
p = 42
# data state: (p = 42)
print(p)
```

## Reassigning variables
After assigning a variable in a statement, the variable can be reassigned a new value of any type in a later statement.
When a variable is assigned to new value, it simply replaces the old value in the storage space.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-3.jpg"
alt="Reassigning variables" width="200"  height="auto">

```python
p = 42
print(p)
p = 42.0
print(p)
p = True
print(p)
p = "42"
print(p)
```

### Operator assignment shortcuts
It is common to add, substract, multiply, or divide the variable with a value.

```python
p = 40
p = p + 2
print(p)
q = 50
q = q - 8
print(q)
r = 21
r = r * 2
print(r)
s = 126
s = s / 3
print(s)
s = 128
s = s % 43
print(s)
s = 42
s = s ** 2
print(s)
```

__Exercise__. For `t = "42 "`, compute `t = t * 2` and print it.

Because this is so common, there are shortcut operators that are more concise.
| shortcut | operation |
| - | - |
| += | addition assignment |
| -= | substraction assignment |
| *= | multiplication assignment |
| /= | division assignment |
| %=	| remainder |
| \*\*= | exponentiation|


```python
p = 40
p += 2
print(p)
q = 50
q -= 8
print(q)
r = 21
r *= 2
print(r)
s = 126
s /= 3
print(s)
s = 128
s %= 43
print(s)
s = 42
s **= 2
print(s)
```

## Assigning variables to data structure types
When a variable is assigned to a basic type value: `list`s, `set`s, `dictioaries`s, the new value is not as simple as with basic types.

### Assigning variables to `lists`
A list value can be arbitrarily large, so it cannot necessarily fit in a fixed storage space.
Instead, a list is stored in a separate storage space and the variable is assigned a *reference* to that storage space.
