## Variable names
Variables have unique names.
There are restrictions on variable names:
- The characters in the name are limited to:
| valid characters | examples |
| - | - |
| upper and lower case letters | `A`..`Z` and `a`..`z` |
| numbers | `0`..`9` |
| underscores | `_` |
- They must not start with a number (and avoid starting with an underscore `_`, it's legal but could have special meaning to Python)
- They are case-sensitive so that `Name` and `NAME` are different variables
- They cannot be one of the built-in [reserved keywords](https://docs.python.org/3/reference/lexical_analysis.html#keywords) used for statements or operators

Ideally, variable names should be a reminder of their purpose.
When multiple words describe a variable's function,
customary ways to connect the words are:
- To use an underscore `_` to connect the words (*snake_case*), such as `variable_name`
- Use *camelCase*, such as `variableName`

There are some more notes on variable names:
- It is customary to begin the name for a simple variable with a lowercase letter.
- Sometimes a variable is used for a fixed value, like a conversion factor.
A custom is to use a name with only uppercase and underscores for a constant.


````python
KILOMETERS_TO_MILES = 0.6214
kilometers = 100
miles = kilometers * KILOMETERS_TO_MILES
print(miles)
````

## Assigning variables to basic data types
When assigning a variable to a basic data type value, `int`, `float`, `boolean`, or `string`,
it is given a small place in memory to store a value.
That variable is placed into the data state.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-variables-2.jpg"
alt="Assigning variables" width="200" height="auto">

````python
# data state 0: ()
p = 42
print(p)
# data state 1: (p = 42)
q = 42.0
print(q)
# data state 2: (p = 42, q = 42.0)
r = True
print(r)
# data state 3: (p = 42, q = 42.0, r = True)
s = "42"
print(s)
# data state 4: (p = 42, q = 42.0, r = True, s = "42")
````

Variables are assigned to the value of *expressions*.
An expression is a combination of variables, constants, operators, and functions that produce a value.
Expressions follow the rules of standard mathematical statements.
Any place a variable or function appears in an expression, its value is substituted.

````python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2, height = 21)
area = width * height
# data state: (width = 2, height = 21, area = 42)
print(area)
````

## Reassigning variables
After assigning a variable in a statement, the variable can be reassigned a new value of any type in a later statement.
When a variable is assigned to new value, it simply replaces the old value in the storage space.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-variables-3.jpg"
alt="Reassigning variables" width="300"  height="auto">

````python
# data state 0: ()
p = 42
print(p)
# data state 1: (p = 42)
p = 42.0
print(p)
# data state 2: (p = 42.0)
p = "42"
print(p)
# data state 3: (p = "42")
p = True
print(p)
# data state 4: (p = True)
````

## Multiple variable assignment
As a convenience, more than one variable can be assigned in a single statement:

*variable*`, `*variable*`, ... = `*value*`, `*value*`, ...`

There must be as many variables as values.

````python
p, q, r = 42, 42.0, "42"
print(p, q, r)
````

A useful example is to swap the values of variables.

````python
p, q = 42, 42.0
print(p, q)
p, q = q, r
print(p, q)
````

## Variable `is` operator

````python
a = 1
b = 1
c = a
print(a is b)
print(a == b)
print(a is c)
print(a == c)

````

````python
a = 1000
b = 1000
c = a
print(a is b)
print(a == b)
print(a is c)
print(a == c)
````

````python

list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a  # c points to the same object as a

print(list_a == list_b)  # True  -> Their values are identical
print(list_a is list_b)  # False -> They are stored in different memory locations

print(list_a is list_c)  # True  -> 'list_c' is just a nickname for 'list_a'
````

Best Practice: When should you actually use is?
The most common and "Pythonic" use case for the is operator is checking if a variable is None, True, or False. Because None is a singleton (only one None object ever exists in a running program), is is the safest and fastest way to check it. [4, 6, 7]
```
# The right way to check for None
if user_input is None:
    print("No input provided")
```

### Statements that modify lists
The following are operations that modify lists assigned to variables.

````python
# add a value to a list
a_list = ["string", 1.1, 42]
a_list.append(12)
print(a_list)
````

````python
# add a list to the end of a list
a_list = ["string", 1.1, 42]
a_list.extend([12, "end"])
print(a_list)
````

````python
# remove a value from a list
a_list = ["string", 1.1, 42]
a_list.remove(42)
print(a_list)
````

````python
# add a value at a specific position
a_list = ["string", 1.1, 42]
a_list.insert(0, 3)
print(a_list)
````

````python
# delete a value at a specific position
a_list = ["string", 1.1, 42]
del a_list[1]
print(a_list)
````

````python
# reverse a list
a_list = ["string", 1.1, 42]
a_list.reverse()
print(a_list)
````

````python
# sort a list
a_list = [3, 12.1, 42, 10]
a_list.sort()
print(a_list)
````

````python
# s[i] = x
# item i of s is replaced by x
l = ["1", "2", "3"]
l[1] = "4"
print(l)
````

````python
# del s[i]
# removes item i of s
l = ["1", "2", "3"]
del l[1]
print(l)
````

````python
# s[i:j] = t
# slice of s from i to j is replaced by the contents of the iterable t
l = ["1", "2", "3", "4"]
l[1:2] = "5"
print(l)
l = ["1", "2", "3", "4"]
l[1:3] = "5"
print(l)
l = ["1", "2", "3", "4"]
l[1:2] = ["5", "6"]
print(l)
l = ["1", "2", "3", "4"]
l[1:3] = ["5", "6"]
print(l)
````

````python
# del s[i:j]
# removes the elements of s[i:j] from the list (same as s[i:j] = [])
l = ["1", "2", "3"]
del l[1:2]
print(l)
l = ["1", "2", "3", "4"]
del l[1:2]
print(l)
l = ["1", "2", "3"]
l[1:2] = []
print(l)
l = ["1", "2", "3"]
del l[1:3]
print(l)
l = ["1", "2", "3", "4"]
del l[1:3]
print(l)
l = ["1", "2", "3"]
l[1:3] = []
print(l)
````

````python
# s += t
# extends s with the contents of t (for the most part the same as s[len(s):len(s)] = t)
l = ["1", "2", "3"]
l += "4"
print(l)
l = ["1", "2", "3"]
l += ["4", "5"]
print(l)
````

````python
# s *= n
# updates s with its contents repeated n times
l = ["1", "2", "3"]
l *= 2
print(l)
````

````python
# Append value to the end of the sequence. This is equivalent to writing seq[len(seq):len(seq)] = [value].
print(["1", "2", "3"].append("4"))
l = ["1", "2", "3"]
l.append("4")
print(l)
````

````python
# Remove all items from sequence. This is equivalent to writing del sequence[:].
print(["1", "2", "3"].clear())
l = ["1", "2", "3"]
l.clear()
print(l)
````

````python
# Create a shallow copy of sequence. This is equivalent to writing sequence[:].
print(["1", "2", "3"].copy())
````

````python
# Extend sequence with the contents of iterable. For the most part, this is the same as writing seq[len(seq):len(seq)] = iterable.
print(["1", "2", "3"].extend(["4"]))
l = ["1", "2", "3"]
l.extend(["4"])
print(l)
````

````python
# Retrieve the item at index and also remove it from sequence. By default, the last item in sequence is removed and returned.
print(["1", "2", "4"].insert(2, "3"))
l = ["1", "2", "4"]
l.insert(2, "3")
print(l)
````

````python
# Retrieve the item at index and also remove it from sequence. By default, the last item in sequence is removed and returned.
print(["1", "2", "3"].pop())
l = ["1", "2", "3"]
l.pop()
print(l)
print(["1", "2", "3"].pop(1))
l = ["1", "2", "3"]
l.pop(1)
print(l)
````

````python
# Remove the first item from sequence where sequence[i] == value.
print(["1", "2", "3"].remove("2"))
l = ["1", "2", "3"]
l.remove("2")
print(l)
````

````python
# Reverse the items of sequence in place. This method maintains economy of space when reversing a large sequence.
# To remind users that it operates by side-effect, it returns None.
print(["1", "2", "3"].reverse())
l = ["1", "2", "3"]
l.reverse()
print(l)
````

````python
print(list(reversed(["1", "2", "3"])))
````

````python
# This method sorts the list in place, using only < comparisons between items.
# Exceptions are not suppressed - if any comparison operations fail, the entire sort operation will fail (and the list will likely be left in a partially modified state).
print(["1", "2", "3"].sort())
l = ["1", "2", "3"]
l.sort()
print(l)
l = ["1", "2", "3"]
l.sort(reverse=True)
print(l)
````

### Statements that modify dictionaries
The following are operations that modify dictionaries assigned to variables

````python
# change a values in a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict["apple"] = "yellow"
print(a_dict)
````

````python
# add a value to a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict["earth"] = "brown"
print(a_dict)
````

````python
# remove a value from a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
del a_dict["sky"]
print(a_dict)
````

````python
# combine two dictionaries
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
another_dict = {"apple": "yellow", "sunset": "orange"}
a_dict.update(another_dict)
print(a_dict)
````

````python
# remove all pairs in a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict.clear()
print(a_dict)
````

````python
# Set d[key] to value
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d["Gandalf"] = "Mithrandir"
print(d)
````

````python
# Remove d[key] from d. Raises a KeyError if key is not in the map
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
del d["Gandalf"]
print(d)
````

````python
# Remove all items from the dictionary
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d.clear()
print(d)
````

````python
# Return a shallow copy of the dictionary
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.copy())
````

````python
# Return a new view of the dictionary’s items ((key, value) pairs)
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(list(d.items()))
````

````python
# If key is in the dictionary, remove it and return its value, else return default.
# If default is not given and key is not in the dictionary, a KeyError is raised.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.pop("Gandalf"))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.pop("Gandalf"))
print(d)
````

````python
# If key is in the dictionary, remove it and return its value, else return default.
# If default is not given and key is not in the dictionary, a KeyError is raised.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.pop("missing", "not there"))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.pop("missing", "not there"))
print(d)
````

````python
# Remove and return a (key, value) pair from the dictionary. Pairs are returned in LIFO order.
# popitem() is useful to destructively iterate over a dictionary, as often used in set algorithms.
# If the dictionary is empty, calling popitem() raises a KeyError.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.popitem())
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.popitem())
print(d)
print(d.popitem())
print(d)
````

````python
# Return a reverse iterator over the keys of the dictionary. This is a shortcut for reversed(d.keys())
print(list(reversed({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"})))
````

````python
# Update the dictionary with the key/value pairs from mapping or iterable and kwargs, overwriting existing keys. Return None.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"}.update({"Gimli": "Dwarf"}))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.update({"Gimli": "Dwarf"}))
print(d)
````

````python
# Create a new dictionary with the merged keys and values of d and other, which must both be dictionaries.
# The values of other take priority when d and other share keys.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"} | {"Gimli": "Dwarf"})
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"}
d |= {"Gimli": "Dwarf"}
print(d)
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d |= {"Gandalf": "Mithrandir"}
print(d)
````
