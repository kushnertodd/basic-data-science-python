### Variable assignment statements
The most basic program statement assigns a value to a variable with `=`.
Variables are assigned to the value of *expressions*.
An expression is a combination of programming language variables, constants, operators, and functions that produce a single value.
Expressions follow the rules of standard mathematical statements.
Any place a variable or function appears in an expression, its value is substituted.

A variable initially is empty. It must be assigned a value before it is used, else it gives an error. This commonly occurs when a variable name is misspelled.

```python
# data state: ()
p = 42
# data state: (p, value 42)
print(p)
```

The data state at this point is:
- (p, value 42)

Any variable then can be used and have its value substituted in an expression in any following statement.

```python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2; height = 21)
area = width * height
# data state: (width = 2; height = 21; area = value 42)
print(area)
```

### Reassigning variables
After assigning a variable in a statement, the variable can be assigned a new value in a later statement.

```python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2; height = 21)
area = width * height
print(area)
# data state: (width = 2; height = 21; area = 42)
width = 4
# data state: (width = 4; height = 21; area = 42)
area = width * height
# data state: (width = 4; height = 21; area = 84)
print(area)
```

### Operator assignment shortcuts
A common variable assignment statement is to add to, substracts from, multiply by, or divide a variable with a value.

```python
p = 40
p = p + 2
print(p)
q = 50
q = q - 8
print(q)
r = 21
r = r * 2
print(r)
s = 126
s = s / 3
print(s)
```

Because this is so common, there are shortcut operators that are more concise.
| shortcut | operation |
| - | - |
| += | addition assignment |
| -= | substraction assignment |
| *= | multiplication assignment |
| /= | division assignment |

```python
p = 40
p += 2
print(p)
q = 50
q -= 8
print(q)
r = 21
r *= 2
print(r)
s = 126
s /= 3
print(s)
```

### Assigning variables to basic types
When a variable is assigned to a basic type value: `int`s, `float`s, `string`s, the new value simply replaces the old value.

```python
i = 42
print(i)
i = 43
print(i)
f = 42.0
print(f)
f = 43.0
print(f)
s = "42"
print(s)
s = "43"
print(s)
```

### Assigning variables to data structure types
When a variable is assigned to a basic type value: `list`s, `set`s, `dictioaries`s, the new value is not as simple as with basic types.

A list comprehension puts the `for` loop inside square brackets `[]` of a list where the form is:  
`[`*value* `for` *variable* `in` *list*`]`

```python
squares = [x**2 for x in range(10)]
print(squares)
```
