# Variables
The most basic program statement assigns a value to a variable with `=`.
A variable initially is empty.
When a variable is assigned, it is given a small place in memory to store a value.
Assigning a variable to a basic data type value, `int`, `float`, `boolean`, or `string`, essentially places the value into that memory location.
That variable is placed into the data state.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-2.jpg"
alt="Assigning variables" width="200" height="auto">



```python
# data state 0: ()
p = 42
print(p)
# data state 1: (p = 42)
q = 42.0
print(q)
# data state 2: (p = 42, q = 42.0)
r = True
print(r)
# data state 3: (p = 42, q = 42.0, r = True)
s = "42"
print(s)
# data state 4: (p = 42, q = 42.0, r = True, s = "42")
```

Variables are assigned to the value of *expressions*.
An expression is a combination of variables, constants, operators, and functions that produce a value.
Expressions follow the rules of standard mathematical statements.
Any place a variable or function appears in an expression, its value is substituted.

```python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2, height = 21)
area = width * height
# data state: (width = 2, height = 21, area = 42)
print(area)
```

## Reassigning variables
After assigning a variable in a statement, the variable can be reassigned a new value of any type in a later statement.
When a variable is assigned to new value, it simply replaces the old value in the storage space.

<img src="https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/main/docs/images/diagrams-variables-3.jpg"
alt="Reassigning variables" width="200"  height="auto">

```python
# data state 0: ()
p = 42
print(p)
# data state 1: (p = 42)
p = 42.0
print(p)
# data state 2: (p = 42.0)
p = True
print(p)
# data state 3: (p = True)
p = "42"
print(p)
# data state 4: (p = "42")
```

### Operator assignment shortcuts
It is common to add, substract, multiply, or divide the variable with a value.

```python
p = 40
p = p + 2
print(p)
```

```python
q = 50
q = q - 8
print(q)
```

```python
r = 21
r = r * 2
print(r)
```

```python
s = 126
s = s / 3
print(s)
```

```python
t = 128
t = t % 43
print(t)
```

```python
u = 42
u = u ** 2
print(u)
```

### Multiple variable assignment
As a convenience, more than one variable can be assigned in a single statement:

*variable*`, `*variable*`, ... = `*value*`, `*value*`, ...`

There must be as many variables as values.

```python
p, q, r = 42, 42.0, "42"
print(p, q, r)
```

A useful example is to swap the values of variables.

```python
p, q = 42, 42.0
print(p, q)
p, q = q, r
print(p, q)
```

## Variable names
Variables have unique names.
There are restrictions on variable names:
- The characters in the name are limited to:
| valid characters | examples |
| - | - |
| upper and lower case letters | `A`..`Z` and `a`..`z` |
| numbers | `0`..`9` |
| underscores | `_` |
- They must start with a letter or underscore `_`, not a number
- They are case-sensitive so that `Name` and `NAME` are different variables
- They cannot be one of the built-in [reserved keywords](https://docs.python.org/3/reference/lexical_analysis.html#keywords) used for statements or operators

Ideally, variable names should be a reminder of their purpose.
When multiple words describe a variable's function,
customary ways to connect the words are:
- To use an underscore `_` to connect the words (*snake_case*), such as `variable_name`
- Use *camelCase*, such as `variableName`

Sometimes a variable is used for a fixed value, like a conversion factor.
A custom is to use a name with only uppercase and underscores for a constant.


```python
KILOMETERS_TO_MILES = 0.6214
kilometers = 100
miles = kilometers * KILOMETERS_TO_MILES
print(miles)
```

## Assigning variables to data structure types
When a variable is assigned to a basic type value: `list`s, `set`s, `dictioaries`s, the new value is not as simple as with basic types.

### Assigning variables to `lists`
A list value can be arbitrarily large, so it cannot necessarily fit in a fixed storage space.
Instead, a list is stored in a separate storage space and the variable is assigned a *reference* to that storage space.
