### Assigning variables to data structure types
When a variable is assigned to a basic type value: `list`s, `set`s, `dictioaries`s, the new value is not as simple as with basic types.

### Statements that modify lists
The following are operations that modify lists assigned to variables.

```python
# add a value to a list
a_list = ["string", 1.1, 42]
a_list.append(12)
print(a_list)
```

```python
# add a list to the end of a list
a_list = ["string", 1.1, 42]
a_list.extend([12, "end"])
print(a_list)
```

```python
# remove a value from a list
a_list = ["string", 1.1, 42]
a_list.remove(42)
print(a_list)
```

```python
# add a value at a specific position
a_list = ["string", 1.1, 42]
a_list.insert(0, 3)
print(a_list)
```

```python
# delete a value at a specific position
a_list = ["string", 1.1, 42]
del a_list[1]
print(a_list)
```

```python
# reverse a list
a_list = ["string", 1.1, 42]
a_list.reverse()
print(a_list)
```

```python
# sort a list
a_list = [3, 12.1, 42, 10]
a_list.sort()
print(a_list)
```

### Statements that modify sets
The following are operations that modify sets assigned to variables

```python
# add a value to set
a_set = {"apple", "banana", "orange", "pear"}
a_set.add("grape")
print(a_set)
```

```python
# remove a value from set
a_set = {"apple", "banana", "orange", "pear"}
a_set.remove("orange")
print(a_set)
```

```python
# remove all values from a set
a_set = {"apple", "banana", "orange", "pear"}
a_set.clear()
print(a_set)
```

### Statements that modify dictionaries
The following are operations that modify dictionaries assigned to variables

```python
# change a values in a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict["apple"] = "yellow"
print(a_dict)
```

```python
# add a value to a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict["earth"] = "brown"
print(a_dict)
```

```python
# remove a value from a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
del a_dict["sky"]
print(a_dict)
```

```python
# combine two dictionaries
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
another_dict = {"apple": "yellow", "sunset": "orange"}
a_dict.update(another_dict)
print(a_dict)
```

```python
# remove all pairs in a dictionary
a_dict = {"leaf": "green", "sky": "blue", "apple": "red"}
a_dict.clear()
print(a_dict)
```
