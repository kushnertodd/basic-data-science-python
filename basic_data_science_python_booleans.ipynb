## Booleans

Booleans are truth values, and always have one of the two values True and False.
There are a number of comparison operations that return booleans.
They work for integer, float, as well as string types.
For integers and floats, they work as expected.
For strings, characters are compared from the left to find the result.

| Comparison | Meaning |
| - | - |
| < |  less than |
| <= | less than or equal |
| > |  greater than |
| >= | greater than or equal |
| == | equal |
| != | not equal |


```python
# string less than
"abc" < "abc"
```

```python
# string less than or equal
"abc" <= "abc"
```

```python
# string greater than
"abc" > "abc"
```

```python
# string greater than or equal
"abc" >= "abc"
```

```python
# string equal
"abc" == "abc"
```

```python
# string not equal
"abc" != "abc"
```

There are three operators to change the result of conditions.

| operator | example | result |
| - | - | - |
| `and` | *condition* and *condition* | True if both conditions are true  |
| `or` | *condition* or *condition* | True if one of the conditions is true |
| `not` | not *condition* | Reverses the result, False if the condition is true |

The results of the operators can be summarized in a *truth table*.

| a | b | a `and` b | a `or` b | `not` a |
| - | - | - | - | - |
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

```python
# a string begins with 'a' (true) and is length 3 (true)
("abc"[0] == 'a') and (len("abc") == 3)
```

```python
# a string ends with 'bc' (true) or is longer than 5 (false)
("abc"[-2:] == 'bc') or (len("abc") > 5)
```

```python
# a string is not 2 long (false)
# could also have used "len('abc') != 2"
not (len('abc') == 2)
```

### String Properties
These are nine functions to test string properties:

| Function | Description |
| - | - |
| `startswith()` | Whether a string starts with a given string |
| `endswith()` | Whether a string ends with a given string |
| `isalnum()` | Whether a string contains only letters and digits |
| `isalpha()` | Whether a string contains only letters |
| `isdigit()` | Whether a string contains only digits |
| `islower()` | Whether a string is in lower case |
| `isspace()` | Whether the string contains only whitespace characters |
| `isupper()` | Whether a string is in UPPER CASE |
| `in` | Whether a string is in a given string |
| `not in` | Whether a string is not in a given string |



#### `startswith()`
Whether a string starts with a given substring.

```python
# test a string starts with a substring
'Python'.startswith('Pyth')
```

```python
# test a string starts with a substring, starting at a  position
'Python'.startswith('Pyth', 2)
```

#### `endswith()`
Whether a string ends with a given substring.

```python
# test a string ends with a substring
'Python'.endswith('thon')
```

```python
# test a string ends with a substring, starting at a position
'Python'.endswith('thon', 3)
```

#### `isalnum()`
Whether a string contains only letters and digits.

```python
# test if all characters in the string are alphanumeric
'abc123'.isalnum()
```

#### `isalpha()`
Whether a string contains only letters.

```python
# test if all characters in the string are alphabetic
'abcdef'.isalpha()
```

#### `isdigit()`
Whether a string contains only digits.

```python
# test if all characters in the string are decimal characters (0 to 9)
'0123456789'.digit()
```

#### `islower()`
Whether a string is in lower case.

#### `isspace()`
Whether the string contains only whitespace characters (space and tab).

```python
# test if all characters in the string are spaces or tabs
'   '.isspace()
```

#### `isupper()`
Whether a string is in UPPER CASE.

```python
# test if all characters in the string are uppercase
'BANANA'.isupper()
```

#### `in`
Whether a substring is in a given string.

```python
# test if a substring is in a string
'ytho' in 'Python'
```

#### `not in`
Whether a substring is not in a given string.

```python
# test if a substring is not in a string
'ython' not in 'Python'
```
