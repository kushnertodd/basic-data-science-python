## Booleans

Booleans are truth values, and always have one of the two values True and False.
There are a number of comparison operations that return booleans.
They work for integer, float, as well as string types.
For integers and floats, they work as expected.
For strings, characters are compared from the left to find the result.

| Comparison | Meaning |
| - | - |
| < |  less than |
| <= | less than or equal |
| > |  greater than |
| >= | greater than or equal |
| == | equal |
| != | not equal |


```python
# string less than
print("abc" < "abc")
```

```python
# string less than or equal
print("abc" <= "abc")
```

```python
# string greater than
print("abc" > "abc")
```

```python
# string greater than or equal
print("abc" >= "abc")
```

```python
# string equal
print("abc" == "abc")
```

```python
# string not equal
print("abc" != "abc")
```

There are three operators to change the result of conditions.

| operator | example | result |
| - | - | - |
| `and` | *condition* and *condition* | True if both conditions are true  |
| `or` | *condition* or *condition* | True if one of the conditions is true |
| `not` | not *condition* | Reverses the result, False if the condition is true |

The results of the operators can be summarized in a *truth table*.

| a | b | a `and` b | a `or` b | `not` a |
| - | - | - | - | - |
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

```python
# True and True
print("abc" < "def" and "abc" != "def")
```

### String Properties
These are nine functions to test string properties:

| Function | Description |
| - | - |
| `startswith()` | Whether a string starts with a given string |
| `endswith()` | Whether a string ends with a given string |
| `isalnum()` | Whether a string contains only letters and digits |
| `isalpha()` | Whether a string contains only letters |
| `isdigit()` | Whether a string contains only digits |
| `islower()` | Whether a string is in lower case |
| `isspace()` | Whether the string contains only whitespace characters |
| `isupper()` | Whether a string is in UPPER CASE |
| `in` | Whether a string is in a given string |
| `not in` | Whether a string is not in a given string |



### `startswith()` function
There are three ways to use `startswith()`:

| use | meaning |
| - | - |
| *string*.`startswith(`*substring*`)` | whether a string starts with substring |
| *string*.`startswith(`*substring*`, `*start*`)` | whether a string starts with substring beginning at *start* |
| *string*.`startswith(`*substring*`, `*start*`, `*end*`)` |  whether a string starts with substring between *start* and *end* |

```python
# test whether it starts with "DON'T"
print("DON'T PANIC!".startswith("DON'T"))
```

```python
# test whether it starts with a "PANIC!" in the middle
print("DON'T PANIC!".startswith("PANIC!", 6))
```

```python
# test whether it starts with a "PANIC!" in the middle when the string is truncated
print("DON'T PANIC!".startswith("PANIC!", 6, 11))
```

### `endswith()` function
There are three ways to use `endswith()`:

| use | meaning |
| - | - |
| *string*.`endswith(`*substring*`)` | whether a string ends with substring |
| *string*.`endswith(`*substring*`, `*start*`)` | whether a string ends with substring beginning at *start* |
| *string*.`endswith(`*substring*`, `*start*`, `*end*`)` |  whether a string ends with substring between *start* and *end* |

```python
# test a string ends with "PANIC!"
print("DON'T PANIC!".endswith("PANIC!"))
```

```python
# test a string ends with "PANIC!" when the string is truncated
print("DON'T PANIC!".endswith("PANIC!", 10))
```

```python
# test a string ends with "DON'T" at the beginning of the string
print("DON'T PANIC!".endswith("DON'T", 0, 5))
```
