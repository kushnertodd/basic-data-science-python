## Dictionaries
Dictionaries are groups of entries where each entry has a *key* that is used to look up a value, like looking up a word definition in a real dictionary.
Dictionary entries are unordered.
Dictionary keys can be any basic type and values are of any type.
These are covered in the following topics.

| Topic | Description |
| - | - |
| Creating dictionaries | Building dictionaries from sequences of entries |
| Accessing entries | Extracting entries from dictionaries |
| Searching dictionaries | Finding entries in dictionaries |
| Dictionary Properties | Obtaining dictionary information |
| Modifying dictionaries | Changing dictionary content |



## Creating Dictionaries
There are two ways to construct dictionaries.

| method | description |
| - | - |
| braces| entries between braces `{}` |
| `dict()` function | constructs dictionaries |

### Braces
Dictionaries contain entries between braces `{}`, where each entry is a key followed by a colon `:` and a value.
Dictionaries can be used to describe a series of items, such as the characters from [Lord of the Rings](https://www.outfit4events.com/eur/articles/online-store/who-s-who-in-the-lord-of-the-rings-characters-and-their-stories/):
```
{
    "Frodo " : "Hobbit",
    "Gandalf" : "Wizard",
    "Legolas" : "Elf",
    "Gimli" : "Dwarf"
}
```
This could be visualized as:
| character | tribe |
| - | - |
| "Frodo " | "Hobbit" |
| "Gandalf" | "Wizard" |
| "Legolas" | "Elf" |
| "Gimli" | "Dwarf" |

Also, dictionaries can be used to describe attributes of a single item:
```
{
  "company" : "Google",
  "street" : "1600 Amphitheatre Parkway",
  "city" :  "Mountain View",
  "state" : "CA",
  "zip" : 94043,
  "products" : ["Gmail", "Chrome"]
}
```
This could be visualized as:
| attribute | value |
| - | - |
| "company" | "Google" |
| "street" | "1600 Amphitheatre Parkway" |
| "city" |  "Mountain View" |
| "state" | "CA" |
| "zip" | 94043 |
| "products" | ["Gmail", "Chrome"] |

````python
print({
    "Frodo" : "Hobbit",
    "Gandalf" : "Wizard",
    "Legolas" : "Elf",
    "Gimli" : "Dwarf"
})
````

### `dict()` function
There are three ways to use `dict()`:

| function | description |
| - | - |
| `dict(`*key*`=`*value*`, `*key*`=`*value*`, ...)` | create with a list of *key*`=`*value* pairs |
| `dict((`*key*`, `*value*`), (`*key*`, `*value*`), ...)` | create with a list of `(`*key*`, `*value*`)` pairs|
| `dict(zip(`*key-list*`, `*value-list*`))` | using the `zip()` to pair entries from a *key-list* and a *value-list* |

#### `dict(`*key*`=`*value*`, `*key*`=`*value*`, ...)`
Create a dictionary with a list of *key*`=`*value* pairs.

````python
print(dict(Frodo="Hobbit", Gandalf="Wizard", Legolas="Elf", Gimli="Dwarf"))
````

#### `dict((`*key*`, `*value*`), (`*key*`, `*value*`), ...)`
Create a dictionary with a list of `(`*key*`, `*value*`)` pairs.

````python
print(dict([("Frodo", "Hobbit"), ("Gandalf", "Wizard"), ("Legolas", "Elf"), ("Gimli", "Dwarf")]))
````

#### `dict(zip(`*key-list*`, `*value-list*`))`
Create a dictionary using the `zip()` function to pair entries from a *key-list* and a *value-list*.

````python
print(dict(zip(["Frodo", "Gandalf", "Legolas", "Gimli"], ["Hobbit", "Wizard", "Elf", "Dwarf"])))
````

### Accessing entries
Dictionary entry values can be accessed by key using square brackets `[]`.
There are two ways to access dictionary values.

| method | description |
| - | - |
| `[`*key*`]` | the value associated with *key* |
| `get(`*key*`)` | the value associated with *key*  |

````python
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}["Frodo"])
````

## Searching dictionaries
There are two operators to find items in a dictionary.

| method | description |
| - | - |
| `in` | whether an key is in a dictionary |
| `not in` | whether an key isn't in a dictionary |

### *item* in *dictionary*
Whether an key is in a dictionary.

````python
# test if a key is in list
print("Frodo" in {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"})
````

### *item* `not in` *list*
Whether a key is not in a list.

````python
# test if a key is not in list
print("Elrond" in {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"})
````

The `len(`*dictionary*`)` function counts the number of entries in a dictionary.


````python
print(len({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}))
````

## Modifying dictionaries
The `list(`*dictionary*`)` function creates a list of the dictionary keys.
There are two functions that change the contents of a dictionary.

| method | description |
| - | - |
|	`list(`*dictionary*`)` | a list of the dictionary keys |
| `list(`*dictionary*`.keys())` | a list of the dictionary keys |
| `list(`*dictionary*`.values())` | a list of the dictionary values |

````python
print(list({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}))
````

- __Exercise__. Print the keys of `{"Frodo": "Hobbit", "Gandalf": "Wizard"}

````python
print(list({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.keys()))
````

- __Exercise__. Print the keys of `{"Frodo": "Hobbit", "Gandalf": "Wizard"}`.

````python
print(list({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.values()))
````
