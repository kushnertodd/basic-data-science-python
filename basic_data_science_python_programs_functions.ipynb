## Defining Functions
Functions collect blocks of statements that have one purpose.
Python has a variety of built-in functions such as `print()` that we used above.
You can define custom functions for any tasks you have that can make programs clearer and avoid repeating code.
Functions are defined with `def` followed by the function name and parentheses, ending with a colon `:`.
Then the statements of the function follow as an indented block as for conditional statements and loops.

```python
# print a greeting
def hello():
    print("hello")
hello()
```

Functions can vary their processing using *parameters* passed when the functions is called.
The function call must specify *arguments* in the call to the function that matches all parameters defined.

```python
# print the cube of a number
def cube(n):
    print(n*n*n)
cube(8)
```

Functions can return values with the `return` statement.
When the return statement is processed, the value following it is returned as the function value.

```python
def cube(n):
    return n*n*n
print(cube(8))
```

#### Default Parameter Values
When a function is commonly called with a specific value for a parameter, a *default* value may be given to that parameter and the argument can be skipped in the call.

```python
# print a greeting
def hello(greeting="hello"):
    print(greeting)
hello("greetings")
hello()
```

#### Keyword Parameters
Usually function arguments must be in the same order as the parameters where the function was defined.
The arguments can include the parameter name as a *keyword* in the call to the function so that the arguments can be in any order.

```python
# print a greeting with a name
def hello(greeting, name):
    print(greeting, name)
hello("greetings", "earthling")
hello(name="earthling", greeting="greetings")
```

### List Comprehensions
List *comprehensions* provide a concise way to create lists. It is common to produce lists when each element is the result of some operations applied to a sequence of values.

One way to create a list of squares with a `for` loop.

```python
# create a list of squares
squares = []
for x in range(10):
    squares.append(x**2)
print(squares)
```

A list comprehension puts the `for` loop inside square brackets `[]` of a list where the form is:  
`[`*value* `for` *variable* `in` *list*`]`

```python
squares = [x**2 for x in range(10)]
print(squares)
```
