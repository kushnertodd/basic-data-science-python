# Running Python on your own
There are a couple of options for running Python programs once you are ready.

## Running from the web
These are a few websites you can use to run Python interactively from the browser. Most of these are only sandboxes to test small bits of Python code. To develop useful Python programs you will probably need one of the options to run it locally below.

- [CodeChef](https://www.codechef.com/ide)
- [CompilerExplorer](https://godbolt.org)
- [OneCompiler](https://onecompiler.com/python)
- [Online Python](https://www.online-python.com/)
- [Try It Online](https://tio.run/#python3)
- [W3schools](https://www.w3schools.com/tryit/trycompiler.asp?filename=demo_python)

## Google Colab

Google Colab actually lets you create and store Python programs on the web, though because it runs remotely it is more limited for accessing data and less direct for producing production programs.  You can launch it from this website:

https://colab.research.google.com/

Google Colab projects allows you to create Python programs as [Jupyter](https://jupyter.org/) *notebooks*, with explanations combined with Python code. Jupyter is a standard for Data Science projects. This document was produced as a Google Colab project:

[Yet Another Data Science Python Tutorial](https://colab.research.google.com/github/kushnertodd/basic-data-science/blob/main/basic_data_science_python.ipynb)

There are instructions for running it on this webpage: [Basic Data Science: Google Colab](https://basicdatascience.substack.com/p/basic-data-science-google-colab). You can set up a Colab project and it will store the notebook on your Google Drive. For production Python projects where you have more control over the data you read and produce, you probably want to run it from your PC.

## Running from an PC application
The best option for running Python locally is using a PC application that is an Interactive Development Environment (IDE) for running Python.
These provide a user interface to create, maintain, and run Python programs
There is not much reason for using anything else for Windows or MacOS, though these only run on those environments,

### JetBrains PyCharm
PyCharm is the most complete option for running Python on the PC. You can set up the PyCharm environment as follows:
- Go to the PyCharm download page: https://www.jetbrains.com/pycharm/download
- Select Download and run the installer
- To run, launch PyCharm from the Start menu
You will find yourself at a main screen where you can create or open a Python program and run it interactively.
It lets you set up Python projects and manage multiple Python programs simultaneously.
It has help for instructions for how to use the interface.

### Visual Studio Code (Windows)
Visual Studio Code is Microsoft's version of a Python environment.
It actually supports a number of programming languages so is not specialized to Python.
You can set up the environment like this:
- Go to the VS Code download page: https://code.visualstudio.com/download
- Select Download and run the installer
- Alternatively, and probably easier, install Visual Studio Code from the Microsoft Store (latest version)
- To run, launch VS Code from the Start menu
When you create a project, you have to specify it is for Python.
The instructions can help you create and store programs.

## Running from the command line
You may find running Python programs from the command line is more generally available,
though it would require skill in using computer terminal apps.
Python programs can be run from the command line like this:
```
python program.py
```
Programs can then access data on the local filesystem which you can manage with PC tools.
Running Python programs in production can be easier using command line scripts rather than having to run an IDE, load the program, run from the interace, etc.

Here are install instructions for Windows and MacOS.

### Install on Windows
- Go to the Python download page: phttps://www.python.org/downloads/
- Select the latest Python 3 version
- Run the installer and select Add Python to PATH (e.g., Python 3.14)
- Alternatively, and probably easier, install from the Microsoft Store (latest version)
- Open a command prompt and run this to test the installation succeeded:
```
C:> python --version
Python 3.14
```
- An alternative is to launch the Python Shell by running IDLE from the Start menu

### Install on MacOS
- Open Applications > Utilities > Terminal
- Run:
```
% brew install python3
```
- Run this to test the installation:
```
% python --version
Python 3.14
```
