# Numbers
Python has two types of numbers:
| type | name | description |
| - | - | - |
| Integers | `int` | negative and positive whole numbers, without fractions |
| Floating | `float` | continuous real numbers with fractions |




## Integer numbers
Integers, or *int*s, are negative and positive whole numbers, without fractions.
Integers are any sequences of digits optionally preceded by a minus sign.

```python
print(42)
```

Integers can include underscores `_` to group digits to make them more readable.

```python
print(100_000_000_000)
print(100000000000)
```

## Floating-point numbers
Floating-point numbers, or *float*s, are continuous real numbers with fractions.
They are sequences of digits optionally preceded by a minus sign, with a period that starts a fraction.
There may be no digits before or after the period.

```python
print(42.0)
```

```python
print(-42.)
```

```python
print(0.42)
```

```python
print(.42)
```

Floats can also be entered using *exponential* notation,
where there is an `e` and a number *n* at the end:

*float*`e`*n*

The number's value is *float* multiplied by 10 to the *n*:
$$float*10^n$$

```python
print(1000.0)
```

```python
print(1e3)
```

```python
print(42.01)
```

```python
print(4.201e1)
```

## Operations on Numbers
These are operations on numbers.
Python automatically changes the result based on whether the values are integer or floating.

| Expression | Operation | Result |
| - | - | - |
| x + y | addition | integer if both are integers, else floating |
| x - y | subtraction | integer if both are integers, else floating |
| x * y | multiplication | integer if both are integers, else floating |
| x / y | division | always floating |
| x // y | integer division | divide and drop fraction, integer if both are integers, else floating |
| x % y | remainder | integer if both are integers, else floating |
| -x | negation | integer if x is integer, else floating |
| +x | unchanged | integer if x is integer, else floating |
| x ** y | exponentiation | integer if both are integers, else floating |
| pow(x, y) | exponentiation | integer if both are integers, else floating |
| abs(x) | absolute value | integer if x is integer, else floating |
| int(x) | convert to integer | integer for x integer, floating, or string |
| float(x) | convert to floating  | floating for x integer, floating, or string |
| str(x) | convert to string  | string for x integer or floating |
| round() | round | round a floating value |

These are examples of integer operations.

```python
# addition
print(17 + 25)
```

```python
# subtraction
print(66 - 24)
```

```python
# multiplication
print(2 * 21)
```

```python
# division
print(128 // 3)
```

```python
# remainder
print(128 % 43)
```

```python
# exponentiation
print(42 ** 2)
```

```python
# exponentiation
print(pow(42, 2))
```

```python
# absolute value
print(abs(-42))
```

```python
# convert float to integer
print(int(42.0))
```

```python
# convert string to integer
print(int("42"))
```

```python
# convert an integer to a string
print(str(42))
```

These are examples of floating operations.

```python
# addition
print(17.0 + 25)
```

```python
# subtraction
print(66.0 - 24)
```

```python
# multiplication
print(2 * 21.0)
```

```python
# division
print(126 // 3)
```

```python
# remainder
print(128.0 % 43)
```

```python
# exponentiation (square root)
print(1764 ** 0.5)
```

```python
# exponentiation (square root)
print(pow(1764, 0.5))
```

```python
# absolute value
print(abs(-42.0))
```

```python
# convert integer to float
print(float(42))
```

```python
# convert string to float
print(float("42.0"))
```

```python
# convert an float to a string
print(str(42.0))
```

#### `round()`
There are two ways to use `round()`:

| use | meaning |
| - | - |
| `round(`*x*`)` | Round *x* |
| `round(`*x*`, `*digits*`)` |  Round *x* to the given number of digits |

```python
# round up
print(round(41.5))
```

```python
# round down
print(round(42.2))
```

```python
# round to 2 digits
print(round(3.14159, 2))
```
