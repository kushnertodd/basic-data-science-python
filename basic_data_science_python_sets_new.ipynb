### `set()` function
There are two ways to use `set()`:

| use | meaning |
| - | - |
| [`set()`](#scrollTo=set_function) |  the empty set |
| [`set(`*list*`)`](#scrollTo=set_list_function) |  a set of the items in *list* |
| [`set(`*string*`)`](#scrollTo=set_string_function) | a set of the characters in *string* |

#### `set()` function
The empty set (displayed as "set()" instead of "{}").

```python
print(set())
```

#### `set(`*list*`)` function
Creates a set of the items in *list*. Duplicates are removed.


```python
print(set(["apple", "banana", "cherry", "apple"]))
```

#### `set(`*string*`)` function
Creates a set of the characters in *string*. Duplicates are removed.

```python
print(set("cherry"))
```

## Set Properties
There are eight ways to obtain set information.

| Method | Shortcut | Description |
| - | - | - |
| [`<=`](#scrollTo=subset_operator) | whether all items of one set are in another |
| [`<`](#scrollTo=proper_subset_operator) | whether all items of one set are in another (and are not equal) |
| [`>=`](#scrollTo=superset_operator) | whether all items of one set include all those in another |
| [`>`](#scrollTo=proper_superset_operator) | whether all items of one set include all those in another (and are not equal) |
| [`isdisjoint()`](#scrollTo=set_isdisjoint_set_function) | whether two sets have no intersection |
| [`len()`](#scrollTo=len_set_function) | number of set items |
| [`max()`](#scrollTo=max_set_function) | largest set item |
| [`min()`](#scrollTo=min_set_function) | smallest set |

```python
{"banana", "pear"} <= {"apple", "banana", "orange", "pear"}
```

## Set functions
Functions on sets.

| Function | Shortcut | Description |
| - | - | - |
| [difference()](#scrollTo=set_difference_set_function_shortcut_) | - | the items in one set but not in another |
| [intersection()](#scrollTo=set_intersection_set_function_shortcut_) | & | the items in both of two sets  |
| [symmetric_difference()](#scrollTo=set_symmetric_difference_set_function_shortcut_) | ^ | the items not in either of two sets |
| [union()](#scrollTo=set_union_set_set_function_shortcut_) | \| | all unique items on combining two sets |
| [sorted()](#scrollTo=sorted_set_function) | \| | sorted list of all set items |

### *set*.`difference(`*set*`)` function (shortcut: `-`)
The items in one set but not in another.





```python
print({"apple", "orange", "cherry"}.difference({"cherry", "pear", "apple"}))
```

```python
print({"apple", "orange", "cherry"} - {"cherry", "pear", "apple"})
```

- __Exercise__. Find the difference of `{"apple", "orange", "cherry"}` and ` {"grape", "pear"}`

### *set*.`intersection(`*set*`)` function (shortcut: `&`)
The items in both of two sets .


```python
print({"apple", "orange", "cherry"}.intersection({"cherry", "pear", "apple"}))
```

```python
print({"apple", "orange", "cherry"} & {"cherry", "pear", "apple"})
```

```python
print({"apple", "orange", "cherry"}.intersection({"cherry", "pear", "apple"}, {"cherry", "grape"}))
```

```python
print({"apple", "orange", "cherry"} & {"cherry", "pear", "apple"} & {"cherry", "grape"})
```

- __Exercise__. find the intersection of `{"apple", "orange", "cherry"}` and `{"pear", "banana"}`

### *set*.`symmetric_difference(`*set*`)` function (shortcut: `^`)
The items not in either of two sets.


```python
print({"apple", "orange", "cherry"}.symmetric_difference({"cherry", "pear", "apple"}))
```

```python
print({"apple", "orange", "cherry"} ^ {"cherry", "pear", "apple"})
```

- __Exercise__. Find the symmetric difference of `{"apple", "orange", "cherry"}` and `{"apple", "orange", "cherry"}`

### *set*.`union(`*set*, *set*, `...)` function (shortcut: "|")
All unique items on combining sets.
Multiple sets can be combined.

```python
print({"apple", "orange", "cherry"}.union({"cherry", "pear", "apple"}))
```

```python
print({"apple", "orange", "cherry"} | {"cherry", "pear", "apple"})
```

```python
print({"apple", "orange", "cherry"}.union({"cherry", "pear", "apple"}, {"cherry", "grape"}))
```

```python
print({"apple", "orange", "cherry"} | {"cherry", "pear", "apple"} | {"cherry", "grape"})
```

- __Exercise__. Find the union of `{"apple", "orange", "cherry"}` and `{"banana", "pear"}`

### `sorted(`*set*`)` function
A sorted list of all set items.

```python
# sort set
print(sorted({"orange", "grape", "apple", "cherry", "pear"}))
```

- __Exercise__. Sort the set `{"2", "4", "1", "2", "3"}`

Update the set, adding elements from all others.

```python
print({"orange", "grape", "apple",  "pear"}.update({"cherry", "apple"}))
s = {"orange", "grape", "apple",  "pear"}
s.update({"cherry", "apple"})
print(s)
```

```python
s = {"orange", "grape", "apple",  "pear"}
s |= {"cherry", "apple"}
print(s)
```

Update the set, keeping only elements found in it and all others.

```python
print({"orange", "grape", "apple",  "pear"}.intersection_update({"cherry", "apple"}))
s = {"orange", "grape", "apple",  "pear"}
s.intersection_update({"cherry", "apple"})
print(s)
```

```python
s = {"orange", "grape", "apple",  "pear"}
s &= {"cherry", "apple"}
print(s)
```

Update the set, removing elements found in others.

```python
print({"orange", "grape", "apple",  "pear"}.difference_update({"cherry", "apple"}))
s = {"orange", "grape", "apple",  "pear"}
s.difference_update({"cherry", "apple"})
print(s)
```

```python
s = {"orange", "grape", "apple",  "pear"}
s -= {"cherry", "apple"}
print(s)
```

Update the set, keeping only elements found in either set, but not in both.

```python
print({"orange", "grape", "apple",  "pear"}.symmetric_difference_update({"cherry", "apple"}))
s = {"orange", "grape", "apple",  "pear"}
s.symmetric_difference_update({"cherry", "apple"})
print(s)
```

```python
s = {"orange", "grape", "apple",  "pear"}
s ^= {"cherry", "apple"}
print(s)
```

```python
# Add element elem to the set.
print({"orange", "grape", "pear"}.add("apple"))
s = {"orange", "grape", "pear"}
s.add("apple")
print(s)
print({"orange", "grape", "apple",  "pear"}.add("apple"))
s = {"orange", "grape", "apple",  "pear"}
s.add("apple")
print(s)
```

```python
# Remove element elem from the set. Raises KeyError if elem is not contained in the set.
print({"orange", "grape", "apple",  "pear"}.remove("apple"))
s = {"orange", "grape", "apple",  "pear"}
s.remove("apple")
print(s)
```

```python
# Remove element elem from the set if it is present.
print({"orange", "grape", "apple",  "pear"}.discard("apple"))
s = {"orange", "grape", "apple",  "pear"}
s.discard("apple")
print(s)
s = {"orange", "grape", "pear"}
s.discard("apple")
print(s)
```

```python
# Remove and return an arbitrary element from the set. Raises KeyError if the set is empty.
print({"orange", "grape", "apple",  "pear"}.pop())
s = {"orange", "grape", "pear"}
print(s.pop())
print(s)
```

```python
# Remove all elements from the set.
print({"orange", "grape", "apple",  "pear"}.clear())
s = {"orange", "grape", "apple",  "pear"}
s.clear()
print(s)
```
