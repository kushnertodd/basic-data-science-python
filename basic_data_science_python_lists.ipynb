## Lists
Lists are ordered dynamic sequences of values.
List values can be accessed using a position inside square brackets `[]`,
with position 0 being the first value.
Values can be accessed from the end of the list with negative positions.
This is the list `["text", 1, True, 4.2]` with value positions:

```python
# a sample list with mixed types
['text', 42, True, 2.5]
```

```python
# the first value in the list
['text', 42, True, 2.5][0]
```

```python
# the fourth value in the list
['text', 42, True, 2.5][3]
```

```python
# the last value in the list
['text', 42, True, 2.5][-1]
```

```python
# the third to last value in the list
['text', 42, True, 2.5][-3]
```

List *slices* can be accessed by referencing two positions in square brackets separated by a colon `:`.
The values are selected up to, but not including, the second position.

```python
# the second value up to, but not including, the fourth value in the list
['text', 42, True, 2.5][1:3]
```

Slices can include negative positions. Again, the values are selected up to, but not including, the second position.

```python
# the third value to, but not including, the last value in the list
['text', 42, True, 2.5][-3:-1]
```

Either the first or second position may be omitted.
- If the first position is omitted, the slice starts at the beginning of the list, up to but not including the second position.
- If the second position is omitted, the slice starts at the first value and ends with the last value of the list.

```python
# with the first position omitted, the first two values in the list
['text', 42, True, 2.5][:2]
```

```python
# with the second position omitted, the last three values in the list
['text', 42, True, 2.5][3:]
```

```python
# with the first position omitted and a negative second position, the first three values in the list
['text', 42, True, 2.5][:-3]
```

```python
# with a negative first position and omitting the second position, the last four values in the list
['text', 42, True, 2.5][-4:]
```

```python
# select every other list value
['text', 42, True, 2.5][0:3:2]
```

```python
# the length of a list
len(['text', 1, True, 4.2])
```

```python
# find index of first list value
['a', 'b', 'c', 'd', 'b'].index('b')
```

```python
# count instances of a list value
['a', 'b', 'c', 'd', 'b'].count('b')
```

```python
# test if a value is in list
'b' in ['a', 'b', 'c', 'd']
```

```python
# test if a value is not in list
'e' not in ['a', 'b', 'c', 'd']
```

```python
# repeat a list
['a', 'b', 'c', 'd']*2
```

```python
# combine two lists
['a', 'b', 'c', 'd'] + ['e', 'f', 'g', 'h']
```

```python
# list minimum value
min([210.1, 81, 42, 115])
```

```python
# list maximum value
max([20.1, 8, 42, 15])
```

```python
# sum of list values
sum([12, 9, 13, 8])
```

```python
# sorted list
sorted([20.1, 1, 42, 15])
```

```python
# split a string at spaces into a list
'1 2   3'.split()
```

```python
# split a string at a character into a list
'1,2,3'.split(',')
```

```python
# join a list of strings separated by a fill character
"-".join(['one', 'two', 'three'])
```
