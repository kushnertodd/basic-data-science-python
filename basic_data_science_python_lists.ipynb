## Lists
Lists are ordered dynamic sequences of values.
List values can be accessed using a position inside square brackets `[]`,
with position 0 being the first value.
Values can be accessed from the end of the list with negative positions.
These are covered in the following topics.

| Topic | Description |
| - | - |
| Creating lists | Building lists from sequences of items |
| Accessing items | Extracting items from lists |
| Searching lists | Finding substrings in lists |
| List Properties | Obtaining list information |
| Modifying lists | Changing list content |

Lists are constructed in a number of ways.

| method | description |
| - | - |
| square brackets | items between square brackets `[]` |

### Square brackets
Items between square brackets `[]`.

### Functions
There are four functions that construct lists.

| function | description |
| - | - |
| `list()` | create a list with the characters of *string* |
| `split()` | return a list of the words split out from a string |
| `rsplit()` | Return a list of the words in the string, using sep as the delimiter string. |
| `splitlines()` | Return a list of the lines in the string, breaking at line boundaries |

### `list(`*string*`)` function
create a list with the characters of *string*.

```python
print(list("DON'T PANIC!"))
```

### `split()` function
There are three ways to use `split()`:

| use | meaning |
| - | - |
| *string*.`split()` | Return a list of the words in the string separated by whitespace |
| *string*.`split(`*substring*`)` |  Return a list of the words in the string separated by *substring* |
| *string*.`split(`*substring*`, `*maxsplit*`)` | Return a list of at most *maxsplit* words in the string separated by *substring* |

```python
print("DON'T PANIC!".split())
print("DON'T PANIC!".split("N"))
```

- __Exercise__. Split "DON'T PANIC!" at "N" once.

### `rsplit()` function
There are three ways to use `rsplit()`:

| use | meaning |
| - | - |
| *string*.`rsplit()` | Return a list of the words in the string separated by whitespace (same as `split()`) |
| *string*.`rsplit(`*substring*`)` |  Return a list of the words in the string separated by *substring* (same as `split()`) |
| *string*.`split(`*substring*`, `*maxsplit*`)` | Return a list of at most *maxsplit* words in the string separated by *substring*, starting from the right |

```python
print("DON'T PANIC!".rsplit())
print("DON'T PANIC!".rsplit("N"))
print("DON'T PANIC!".rsplit("N", 1))
```

### *string*.`splitlines()` function
create a list of the lines *string* split at newlines (`\n`).

```python
"""DON'T PANIC!
DON'T PANIC!
DON'T PANIC!
""".splitlines()
```

There are two the ways to access items by position.

| Method | Description |
| - | - |
| Indexing lists | Accessing items by single positions |
| List slices | Extracting consecutive items using position ranges |

### Indexing Lists
Individual items can be accessed using a single position inside square brackets `[]`.

```python
# the first value in the list
['text', 42, True, 2.5][0]
```

```python
# the fourth value in the list
['text', 42, True, 2.5][3]
```

```python
# the last value in the list
['text', 42, True, 2.5][-1]
```

```python
# the third to last value in the list
['text', 42, True, 2.5][-3]
```

```python
# the first and last item
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][0])
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][8])
```

### List Slices
List *slices* access sequences of items between positions in square brackets separated by one or more colons `:`.
The ways to use slices are:
- *list*`[`*start*`:`*end* `]`
- *list*`[`*start*`:`*end* `:`*step*`]`

The items are selected from the start position up to, but not including, the end position.
If a step is specified, that many items are skipped for each character selected between the start and end.

```python
# the first four items
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][0:4])

# the last three items
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][5:8])
```

- __Exercise__: Print out the items using negative positions.

```python
# even items
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][0:9:2])
```

- __Exercise__: Print out the even items using negative positions.
- __Exercise__: Print out the odd items.

Either the first or second position may be omitted:
- If the first position is omitted, the slice starts at the beginning of the list, up to but not including the second position.
- If the second position is omitted, the slice starts at the first item and ends with the last character of the list.

```python
# the first five items with the first position omitted
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][:5])

# the last two items with the second position omitted
print(["0", "1", "2", "3", "4", "5", "6", "7", "8"][7:])
```

__Exercise__: print out the items using negative positions.

List *slices* can be accessed by referencing two positions in square brackets separated by a colon `:`.
The values are selected up to, but not including, the second position.

```python
# the second value up to, but not including, the fourth value in the list
['text', 42, True, 2.5][1:3]
```

Slices can include negative positions. Again, the values are selected up to, but not including, the second position.

```python
# the third value to, but not including, the last value in the list
['text', 42, True, 2.5][-3:-1]
```

Either the first or second position may be omitted.
- If the first position is omitted, the slice starts at the beginning of the list, up to but not including the second position.
- If the second position is omitted, the slice starts at the first value and ends with the last value of the list.

```python
# with the first position omitted, the first two values in the list
['text', 42, True, 2.5][:2]
```

```python
# with the second position omitted, the last three values in the list
['text', 42, True, 2.5][3:]
```

```python
# with the first position omitted and a negative second position, the first three values in the list
['text', 42, True, 2.5][:-3]
```

```python
# with a negative first position and omitting the second position, the last four values in the list
['text', 42, True, 2.5][-4:]
```

```python
# select every other list value
['text', 42, True, 2.5][0:3:2]
```

## Searching lists
Finding items in lists

Functions that return lists.
Lists are constructed in a number of ways.

| method | description |
| - | - |
| `in` | whether an item is in a list |
| `not` | whether an item isn't in a list |

### *item* in *list*
Whether an item is in a list.

```python
# test if a value is in list
'b' in ['a', 'b', 'c', 'd']
```

### *item* `not in` *list*
Whether an item isn't in a list.

```python
# test if a value is not in list
'e' not in ['a', 'b', 'c', 'd']
```

## List Properties
Obtaining list information

| method | description |
| - | - |
|	`all()` | Whether all items of the list are true |
|	`any()` | Whether any item of the list is true |
|	`count()` | The number of times an item occurs in a list |
|	`len()` | The number of items in a list |
|	`max()` | The largest item in an list |
|	`min()` | The smallest item in an list |
|	`sum()` | The sums of items in a list |


### `all(`*list*`)`
Whether all items of the list are true.


### `any()`
Whether any item of the list is true.


### *list*.`count(`*item*`)`
The number of times an item occurs in a list.

```python
# count instances of a list item
['a', 'b', 'c', 'd', 'b'].count('b')
```

### `len(`*list*`)`
The number of items in a list.


```python
# the length of a list
len(['text', 1, True, 4.2])
```

### `max(`*list*`)`
The number of items in a list.

```python
# list maximum value
max([20.1, 8, 42, 15])
```

### `min(`*list*`)`
The smallest item in an list.


```python
# list minimum value
min([210.1, 81, 42, 115])
```

### `sum()`
There are two ways to use `sum()`:

| use | meaning |
| - | - |
| *list*.`sum()` | The sums of items in a list |
| *string*.`sum(``*start*`)` |  The sums of items in a list, adding *start* to the sum |

```python
# sum of list values starting with 3
sum([12, 9, 13, 8], 3)
```

Changing list content
