# Placeholders
A dataset is the collection of the data being processed in a data analytics session.
The current values for all that data at any point in time is a *data state*.
The initial data state are all the available raw data.
This data is modified step by step as it is cleaned, analyzed, transformed, and visualized.

Placeholders, or *variables*, are needed to hold the data from step to step.
The placeholders are called *variables* because the data they hold may be changed from step to step.
This is distinct from mathematical variables, whose value never changes once it is associated with them.
The data state are the values contained by all existing variables.

When all steps are completed, the data is no longer needed and is released.
Any data needed for later must be saved before it is removed.

# Programs
A program automates the processing of data through data states, from the initial raw data until the last data state is handled and the session is over.
A program is a recipe and consists of a series of steps where some data is needed for each step, the data is modified, and the changed data results in a new data state. A recipe for making bread gives an example.

- initial data state: (flour, water, and yeast)
- make the dough
  - input: flour, water, and yeast
  - processing: mix and let rise
  - output: raw dough
  - data state: (raw dough)
- knead the dough
  - input: raw dough
  - processing: roll, press, and fold the dough
  - output: smooth dough
  - data state: (smooth dough)
- bake the dough:
  - input: smooth dough
  - processing: heat in the oven
  - output: bread
  - data state: (bread)

The initial data state is (flour, water, yeast) and the final data state is (bread).

In Python, the data state is stored in variables as the types of data and data structures above.
At each step, the data held by a variable may change,
or a variable may be assigned new data values, in which case it no longer needs the previous value and it is released (called *garbage collection*).
The data state is then the new set of values assigned to all variables.

## Program Statements
A program statement is responsible for each step of the processing that changes the data state.
Program statements may:
- assign data to a variable
- modify the data associated with a variable
- pass data to a function that uses the data to cause some external effect such as:
  - saving the data
  - creating a report from the data
  - visualizing the data
  - giving the data to a machine learning algorithm to predict future effects

Each statement modifies the data state.
The progression of a program as it is run is:
```
(initial data state)
program statement 1
(data state 1)
program statement 2
(data state 2)
...
last program statement
(all data is released and the program ends)
```

### Variable assignment statements
The most basic program statement assigns a value to a variable with `=`.
Variables are assigned to the value of *expressions*.
An expression is a combination of programming language variables, constants, operators, and functions that produce a single value.
Expressions follow the rules of standard mathematical statements.
Any place a variable or function appears in an expression, its value is substituted.

A variable initially is empty. It must be assigned a value before it is used, else it gives an error. This commonly occurs when a variable name is misspelled.

````python
# data state: ()
p = 42
# data state: (p, value 42)
print(p)
````

The data state at this point is:
- (p, value 42)

Any variable then can be used and have its value substituted in an expression in any following statement.

````python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2; height = 21)
area = width * height
# data state: (width = 2; height = 21; area = value 42)
print(area)
````

### Reassigning variables
After assigning a variable in a statement, the variable can be assigned a new value in a later statement.

````python
# data state: ()
width = 2
# data state: (width = 2)
height = 21
# data state: (width = 2; height = 21)
area = width * height
print(area)
# data state: (width = 2; height = 21; area = 42)
width = 4
# data state: (width = 4; height = 21; area = 42)
area = width * height
# data state: (width = 4; height = 21; area = 84)
print(area)
````

## Statements that control statement processing order
The most powerful feature of programs is that they can process huge amounts of data and automatically make decisions based on data values.
Statements are normally processed from the top of the program to the end.
There are two statements that select whether a group of statements are to be processed:
- `if` statements that select whether to process a group of statements based on a condition.
- `match` statements that select a group of statements to process based on a value.

There are also two statements that allow processing groups of statements repeatedly:
- `while` statements repeat a group of statements while a condition is true.
- `for` statements that process a group of statements once for each data value in a list.

The statements being controlled by these are grouped by being indented with the exact same indentation for each statement as a *block*.
These statements in each block are then processed from first to last.

### `if` Statements
A block of statements are processed together if a condition is true with an `if` statement.
The condition follows the `if` and a colon `:` ends the line.

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
````

One variation of an `if` statement allows one block of statements to be processed if the condition is true and another to be processed if the condition is false.
The statements to be processed if the condition are false follow an `else` statement that matches an `if` statement that has the same size indentation.
The `else` is followed by a colon `:`.

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
else:
    print('That is not quite the answer we were looking for')
````

Another variation of an `if` statement allows multiple conditions to be tested one by one and process different groups of statements depending on the resuts.If there are multiple conditions to be tested, the first condition is with an `if` statement at the beginning, then rest of the conditions are with an `elif` statement.
The `elif` statement is followed by a colon `:`.
The statements following the `elif` have the same size indentation.
There can be a `else` statement after the last `elif` statement,
though that is not required

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
elif x == 41 or x == 43:
    print('Close, but not quite')
else:
    print('That is not quite the answer we were looking for')
````

There is one more form of `if` statement called a "ternary if".
It is a form of `if` statement that returns a value and is used like this:

*true_value* `if` *condition* `else` *false_value*

where:
- *true_value* is the value if the condition is `True`
- *false_value* is the value if the condition is `False`

````python
the_answer = 42
result = "correct" if the_answer == 42 else "incorrect"
print(result)
````

### `match` Statements
A match statement takes a value and processes the `case` blocks matching the value.
This works if the condition is that an expression is equal to a single value.

````python
word = "abc"
n = len(word)
match n:
    case 1:
        print(word, " has 1 character")
    case 2:
        print(word, " has 2 characters")
    case 3:
        print(word, " has 3 characters")
````

This is like an `elif` statement but is clearer and easier to read.

````python
word = "abc"
n = len(word)
if n == 1:
        print(word, " has 1 character")
elif n == 2:
        print(word, " has 2 characters")
elif n == 3:
        print(word, " has 3 characters")
````

There can be a the last `case` block with the value `_` that acts like an `else` if no other block matches.

````python
word = "abc"
n = len(word)
match n:
    case 1:
        print(word, " has 1 character")
    case 2:
        print(word, " has 2 characters")
    case 3:
        print(word, " has 3 characters")
    case _:
        print(word, " has", n, "characters")
````

### `while` Statements
Statements can be processed as long as a condition is true with the `while` statement.

````python
# print all even numbers up to 10
n = 2
while (n <= 10):
  print(n)
  n += 2
````

### `for` Statements
Statements can be processed for each value in a list with the `for` statement.

````python
# Measure some strings:
words = ['apple', 'pear', 'orange']
for word in words:
    print(word, len(word))
````

A shortcut for a list of a set of numbers is the `range` function.
There are 3 versions of using `range`.
| code | meaning |
| - | - |
| `range(`*n*`)` | create a list from 0 to *n*-1 |
| `range(`*start*`, `*finish*`)` | create a list from *start* to *finish*-1 |
| `range(`*start*`, `*finish*`, `*step*`)` | create a list from *start* to *finish*-1 skipping by *step* |

In the version `range(`*start*`, `*finish*`, `*step*`)`,
the last number will be less than *finish-1* if the last *step* takes it past that value.

````python
# prints numbers from  0 to 4
for i in range(5):
    print(i)
````

````python
# prints numbers from 8 to 11
for i in range(8, 12):
    print(i)
````

````python
# prints odd numbers to 11
for i in range(1, 12, 2):
    print(i)
````

````python
# prints every 3th number under 12
for i in range(3, 12, 3):
    print(i)
````

A `for` loop can use `range` to process each list value using an *index*.

````python
# Measure some strings:
words = ['apple', 'pear', 'orange']
for i in range(len(words)):
    print(i, words[i], len(words[i]))
````

### break and continue Statements
The break statement breaks out of the innermost enclosing for or while loop:

````python
# print squares of numbers until one is divisible by 3
for i in range(2, 10, 2):
  if i % 3 == 0:
     print(i, "is dividible by 3")
     break
  print(i)
````

The continue statement continues with the next iteration of the loop:

````python
# print squares of numbers not divisible by 3
for i in range(2, 10, 2):
  if i % 3 == 0:
      print(i, "is dividible by 3, skipping")
      continue
  else:
      print(i*i)
````

## Defining Functions
Functions collect blocks of statements that have one purpose.
Python has a variety of built-in functions such as `print()` that we used above.
You can define custom functions for any tasks you have that can make programs clearer and avoid repeating code.
Functions are defined with `def` followed by the function name and parentheses, ending with a colon `:`.
Then the statements of the function follow as an indented block as for conditional statements and loops.

````python
# print a greeting
def hello():
    print("hello")
hello()
````

Functions can vary their processing using *parameters* passed when the functions is called.
The function call must specify *arguments* in the call to the function that matches all parameters defined.

````python
# print the cube of a number
def cube(n):
    print(n*n*n)
cube(8)
````

Functions can return values with the `return` statement.
When the return statement is processed, the value following it is returned as the function value.

````python
def cube(n):
    return n*n*n
print(cube(8))
````

#### Default Parameter Values
When a function is commonly called with a specific value for a parameter, a *default* value may be given to that parameter and the argument can be skipped in the call.

````python
# print a greeting
def hello(greeting="hello"):
    print(greeting)
hello("greetings")
hello()
````

#### Keyword Parameters
Usually function arguments must be in the same order as the parameters where the function was defined.
The arguments can include the parameter name as a *keyword* in the call to the function so that the arguments can be in any order.

````python
# print a greeting with a name
def hello(greeting, name):
    print(greeting, name)
hello("greetings", "earthling")
hello(name="earthling", greeting="greetings")
````

### List Comprehensions
List *comprehensions* provide a concise way to create lists. It is common to produce lists when each element is the result of some operations applied to a sequence of values.

One way to create a list of squares with a `for` loop.

````python
# create a list of squares
squares = []
for x in range(10):
    squares.append(x**2)
print(squares)
````

A list comprehension puts the `for` loop inside square brackets `[]` of a list where the form is:  
`[`*value* `for` *variable* `in` *list*`]`

````python
squares = [x**2 for x in range(10)]
print(squares)
````

## A simple Python program to read in data and prepare it for processing

## A Python program to perform data cleaning
Let us put together everything we've learned so far to perform a real data science function: data cleaning of a list of Audible audio books. The format of the data is:

| Column | Breakdown |
| - | - |
| name | Name of the audiobook |
| author | Author of the audiobook |
| narrator | Narrator of the audiobook |
| time | Length of the audiobook |
| releasedate | Release date of the audiobook |
| language | Language of the audiobook |
| stars | No. of stars the audiobook received |
| price | Price of the audiobook in INR |

There is a data cleaning task for a number of fields, which we will discover by looking at the raw data.
For each field that needs cleaning, we will make a decision of how to do so.

````python
import requests

uncleaned_url = "https://raw.githubusercontent.com/kushnertodd/basic-data-science/refs/heads/main/data/Audible%20Dataset/audible_uncleaned.tsv"
uncleaned_response = requests.get(uncleaned_url)
````

````python
'''
given the header fields are:
  ['name', 'author', 'narrator', 'time', 'releasedate', 'language', 'stars', 'price']
the field values for each record can be added to lists created with each of the fieldnames.
example:
  Geronimo Stilton #11 & #12,Writtenby:GeronimoStilton,Narratedby:BillLobely,2 hrs and 20 mins,04-08-08,English,5 out of 5 stars34 ratings,468.00
after splitting on ',' becomes
  ['Geronimo Stilton #11 & #12', 'Writtenby:GeronimoStilton', 'Narratedby:BillLobely', '2 hrs and 20 mins', '04-08-08', 'English', '5 out of 5 stars34 ratings', '468.00']
'''
header = True
uncleaned_names = []
uncleaned_authors = []
uncleaned_narrators = []
uncleaned_times = []
uncleaned_releasedates = []
uncleaned_languages = []
uncleaned_stars = []
uncleaned_prices = []
# Check if the request was successful
if uncleaned_response.status_code == 200:
    uncleaned_data = uncleaned_response.text
    for line in uncleaned_data.splitlines():
      if line.strip(): # line is not blank
        if header:     # skip header line
          header = False
        else:
          uncleaned_audible_fields = line.split('\t')
          uncleaned_names.append(uncleaned_audible_fields[0])
          uncleaned_authors.append(uncleaned_audible_fields[1])
          uncleaned_narrators.append(uncleaned_audible_fields[2])
          uncleaned_times.append(uncleaned_audible_fields[3])
          uncleaned_releasedates.append(uncleaned_audible_fields[4])
          uncleaned_languages.append(uncleaned_audible_fields[5])
          uncleaned_stars.append(uncleaned_audible_fields[6])
          uncleaned_prices.append(uncleaned_audible_fields[7])
    print('names        ', uncleaned_names [0:3])
    print('authors      ', uncleaned_authors [0:3])
    print('narrators    ', uncleaned_narrators [0:3])
    print('times        ', uncleaned_times [0:3])
    print('releasedates ', uncleaned_releasedates [0:3])
    print('languages    ', uncleaned_languages [0:3])
    print('stars        ', uncleaned_stars [0:3])
    print('prices       ', uncleaned_prices [0:3])
````

Now we will parse out each of the above formats into hours and minutes, and convert those into total minutes (60*hours + minutes).
They turn out to reduce to four cases ("[s]" implies an optional "s"):

- "Less than n minute" (n turns out to be '1')
- "*n* hr[s]" (minutes = 0)
- "*n* hr[s] and *n* min[s]"
- "*n* min[s]" (hours = 0)

The optional "s" ('[s]') turns out to match the regular expression `s?`.
We will introduce a custom function, `parse_duration_to_minutes(line)`, to parse out a single uncleaned time field and return the computed total minutes.

````python
import re

def parse_duration_to_minutes(uncleaned_time):
    uncleaned_time = uncleaned_time.strip()

    # 1. Handle special case: "Less than d minute" -> 1 minute
    if re.search(r'Less than \d minute', uncleaned_time):
        return 1

    # 2. Extract hours using 'hrs?' (matches 'hr' or 'hrs')
    hours_match = re.search(r'(\d+)\s*hrs?', uncleaned_time)
    hours = int(hours_match.group(1)) if hours_match else 0

    # 3. Extract minutes using 'mins?' (matches 'min' or 'mins')
    mins_match = re.search(r'(\d+)\s*mins?', uncleaned_time)
    minutes = int(mins_match.group(1)) if mins_match else 0

    # 4. Calculate total minutes
    return (hours * 60) + minutes

cleaned_times = []
for uncleaned_time in uncleaned_times:
  cleaned_time = parse_duration_to_minutes(uncleaned_time)
  cleaned_times.append(cleaned_time)
print('uncleaned_times ', uncleaned_times[0:6])
print('cleaned_times   ', cleaned_times[0:6])
````

It looks like those dates have the format 'day-month-year (2 digit)', where we assume the year is after 2000 (year = 2000 + "year (2 digit)").

Let's put this together.

````python
import re

cleaned_releasedates = []
for uncleaned_releasedate in uncleaned_releasedates:
  # detect dates like 6/11/2020, leave unchanged
  slash_match = re.search(r'(\d+)/(\d+)/(\d+)', uncleaned_releasedate)
  if slash_match:
    cleaned_releasedate = uncleaned_releasedate
    cleaned_releasedates.append(cleaned_releasedate)
  else:
    # detect dates like 30-10-18, convert to 10/30/2018
    hyphen_match = re.search(r'(\d+)-(\d+)-(\d+)', uncleaned_releasedate)
    if hyphen_match:
      hyphendate_field1 = int(hyphen_match.group(1))
      hyphendate_field2 = int(hyphen_match.group(2))
      hyphendate_field3 = int(hyphen_match.group(3))
      cleaned_releasedate = f'{hyphendate_field2}/{hyphendate_field1}/{2000+int(hyphendate_field3)}'
      cleaned_releasedates.append(cleaned_releasedate)
print('uncleaned_releasedates ', uncleaned_releasedates[0:10])
print('cleaned_releasedates ', cleaned_releasedates[0:10])
````
