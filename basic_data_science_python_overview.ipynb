## Data Science

[Data science](https://www.ibm.com/think/topics/data-science) uses mathematics and statistics to enable organizations to extract insights and guide operational decisions from raw data.
It combines mathematics, computing and domain knowledge to solve real-world problems and uncover hidden patterns.
Companies are generating vast amounts of data that data scientists can clean, statistically analyze, transform, and visualize to discover patterns and predict events in the future.
These can be presented as actionable insights to business decision makers.

## Why Python?

Python is currently the most popular  programming language in the world, and is the standard for data science because its simplicity makes it easy to learn and use.
It was created by [Guido van Rossum](https://gvanrossum.github.io/) in 1991, named for the BBC comedy series [Monty Python’s Flying Circus](https://docs.python.org/3/faq/general.html#why-is-it-called-python).
It's usefulness for data science is because of the large ecosystem of open-source third-party packages
and the large community of developers who create libraries and tools to make Python easier to use for Data Science.

# Data
Humans are the third most intelligent species on the planet (after mice and dolphins, as anyone who has read or seen [Hitchhikers Guide to Galaxy](https://hitchhikers.fandom.com/wiki/Dolphins) knows).
Humans are distinguished by their ability to create *symbols*, or abstract signs that mean some concept or object in the world. The oldest human-created object symbols have been found in caves in Indonesia,
including 67,800 year old [handprints](https://www.nytimes.com/2026/01/22/world/asia/indonesia-handprint-rock-art-research.html) and 45,500 year old painted [animals](https://www.nytimes.com/2021/01/13/science/cave-painting-indonesia.html).
In caves in Spain and France, 20,000 year old animal symbols have been found that include [markings](https://www.itv.com/news/tyne-tees/2023-01-05/amateur-archaeologist-helps-crack-ice-age-cave-art-code) that tracked animal mating behavior as an aid to hunters.
The oldest evidence of counting are notches on 44,000 year old bones found in South Africa that may have marked a [lunar cycle](https://afrolegends.com/2019/05/17/the-lebombo-bone-the-oldest-mathematical-artifact-in-the-world/).

Tally marks were a form of concrete counting where symbols formed a one-to-one correspondence with physical objects or concepts. Eventually shepherds used stones to count their flock and [tokens](https://www.thoughtco.com/clay-tokens-mesopotamian-writing-171673) were used in the Middle East to count agricultural goods and livestock.
Over time, the tokens evolved to represent [abstract quantities](https://sites.utexas.edu/dsb/ain-ghazal/tokens-and-writing-the-cognitive-development/) of what was actually counted.
These tokens became marks inscribed on tablets that led to the development of cuneiform as a method of writing letters, words and text.
Alphabets were developed as a way to break words into the individual sounds  that allow writing them with a limited number of symbols.

[Data](https://www.ibm.com/think/topics/data) consists of symbols for descriptive facts, figures, observations, and other useful information.
These can be electronically stored as numbers, letters, and text for automatic processing in the computer.
The purpose of data science is to inspect, clean, transform, and interpret raw data to support data-driven decisions.

# Basic Python Data Types

# Python Data Structures
Python data structures collect data values into a group. These include:
- [Lists](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_lists.ipynb): an ordered sequence of any values
- [Sets](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_sets.ipynb): an unordered group of values with no duplicates
- [Dictionaries](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_dictionaries.ipynb): an unordered group where values are associated with a key
