Now we will parse out each of the above formats into hours and minutes, and convert those into total minutes (60*hours + minutes).
They turn out to reduce to four cases ("[s]" implies an optional "s"):

- "Less than n minute" (n turns out to be '1')
- "*n* hr[s]" (minutes = 0)
- "*n* hr[s] and *n* min[s]"
- "*n* min[s]" (hours = 0)

The optional "s" ('[s]') turns out to match the regular expression `s?`.
We will introduce a custom function, `parse_duration_to_minutes(line)`, to parse out a single uncleaned time field and return the computed total minutes.

````python
import re

def parse_duration_to_minutes(uncleaned_time):
    uncleaned_time = uncleaned_time.strip()

    # 1. Handle special case: "Less than d minute" -> 1 minute
    if re.search(r'Less than \d minute', uncleaned_time):
        return 1

    # 2. Extract hours using 'hrs?' (matches 'hr' or 'hrs')
    hours_match = re.search(r'(\d+)\s*hrs?', uncleaned_time)
    hours = int(hours_match.group(1)) if hours_match else 0

    # 3. Extract minutes using 'mins?' (matches 'min' or 'mins')
    mins_match = re.search(r'(\d+)\s*mins?', uncleaned_time)
    minutes = int(mins_match.group(1)) if mins_match else 0

    # 4. Calculate total minutes
    return (hours * 60) + minutes

cleaned_times = []
for uncleaned_time in uncleaned_times:
  cleaned_time = parse_duration_to_minutes(uncleaned_time)
  cleaned_times.append(cleaned_time)
print('uncleaned_times ', uncleaned_times[0:6])
print('cleaned_times   ', cleaned_times[0:6])
````

## Running from the command line
You may find running Python programs from the command line is more generally available,
though it would require skill in using computer terminal apps.
Python programs can be run from the command line like this:
```
python program.py
```
Programs can then access data on the local filesystem which you can manage with PC tools.
Running Python programs in production can be easier using command line scripts rather than having to run an IDE, load the program, run from the interace, etc.

Here are install instructions for Windows and MacOS.

### Install on Windows
- Go to the Python download page: phttps://www.python.org/downloads/
- Select the latest Python 3 version
- Run the installer and select Add Python to PATH (e.g., Python 3.14)
- Alternatively, and probably easier, install from the Microsoft Store (latest version)
- Open a command prompt and run this to test the installation succeeded:
```
C:> python --version
Python 3.14
```
- An alternative is to launch the Python Shell by running IDLE from the Start menu

### Install on MacOS
- Open Applications > Utilities > Terminal
- Run:
```
% brew install python3
```
- Run this to test the installation:
```
% python --version
Python 3.14
```
