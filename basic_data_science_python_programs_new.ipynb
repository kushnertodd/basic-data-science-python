A program is a recipe and where each statement needs some data, the data is modified, and the changed data results in a new data state.


## Data State
In Python, the data state is stored in *variables* as the basic data types and data structures above.
These are distinct from mathematical variables, whose value never changes once it is associated with them.
At each step, the data held by a variable may change.
The data state is then the new set of values assigned to all variables.

The initial data state is empty.
After the program ends, the final data state is released and any data that is to be retained must be saved before completion.

## Program Statements
A program statement is responsible for each step of the processing that changes the data state.

Program statements may modify the data state with actions such as:
- assigning data to a variable
- modifying the data associated with a variable

They may also call a function that returns a value or causes some external effect such as:
  - saving the data
  - creating a report from the data
  - visualizing the data
  - giving the data to a machine learning algorithm to predict future effects

A data analytics session must be initialized with a *dataset*, that is the collection of data to be processed.
When the initial raw data is read, it is entered into the data state.
This data is modified step by step as it is cleaned, analyzed, transformed, and visualized.

# Python program structure
There are a few components of a Python program:
- Library imports
- Comments
- Statement lines

## Library imports
To use libraries, we use the `import` statement, generally at the top of the program.
It makes the functions in the library available to Python.

`import` *library*

This is an example of importing the standard `os` and `sys` libraries in a program.
```
import os   # built-in operating system library
import sys  # built-in system library
# rest of the program
```

## Comments

Comments may be included to give text descriptions of what the code is doing in two ways:
- Text following `#` on one line is ignored through the end of the line
- Text between two single lines containing three double-quotes (`"""`) or single-quotes (`'''`) is ignored:
```
# this line is ignored

p = 42 # text after the "#" is ignored

“””
This is an ignored multiline comment.
“””
```




## Statement lines
Program statements are listed sequentially in the program.
Long program lines can often be continued on the next line, optionally indented using tabs or spaces.
```
# A program line too long to fit on a page
foo = long_function_name(var_one, var_two,
                         var_three, var_four)
```
This includes statements where line is broken within:
- parentheses `(...)`
- square brackets `[...]`
- braces `{...}`
- triple-quoted strings (`"""` or `'''`)

Occasionally a group of statements must be processed together under some statement, called a *block*.
These lines have exactly equal indentation with spaces or tabs (any size) to indicate processing them together:
```
# two lines run together if a condition is true
if (x < 3 and y == 4):
    z = x + y
    print(x, y, z)
```
Blank lines can be added anywhere between program lines to make it more readable.

## `print()` function

One of the most basic statements is using the `print()` function, which allows printing out the values associated with variables.

There are four ways to use `print()`:

| use | meaning |
| - | - |
| `print(`*expression*`, `*expression*`, ...)` | prints out the values of the expressions separated by a space on a single line (ending with newline `\n`)  |
| `print(`*expression*`, `*expression*`, ..., sep="`*sep-string*`")` |  prints out the values of the expressions separated by *sep-string* on a single line (ending with newline `\n`) |
| `print(`*expression*`, `*expression*`, ..., end="`*end-string*`")` |  prints out the values of the expressions separated by a space on a line ending with *end-string* |
| `print(`*expression*`, `*expression*`, ..., sep="`*sep-string*`", end="`*end-string*`")` |  prints out the values of the expressions separated by *sep-string* on a line ending with *end-string* |

You can use `end=""` to continue two `print()` statements on the same line.

````python
print("DON'T", "PANIC!")
````

## A note on Data State and Google Colab Notebooks
Google Colab notebooks, and all Jupyter based notebooks, remember the data state after running code blocks.
When a code block is run, the data state after any variable assignments or other statements that change data state is remembered when the next code block is run.

In most notebooks it is inconvenient to make each code block an independent Python program, so they expect each code block will be run from the top of the notebook in order.
Later code blocks will depend on the data state changes made by previous blocks.

The key will be when a variable is used in a code block without having been initialized in the block.
If the variable has never been initialized, it will give an error.
‘Run all’ will run all code blocks in order, so avoid this problem.

These three code blocks show this behavior.
- If you run Block 1 before this block, Block 3 prints 1.
- If you run Block 2 before this block, Block 3 prints 2.
- If you don't run either of Block 1 or Block 2 before Block 3, you get an error.

````python
# Block 1
a = 1
````

````python
# Block 2
a = 2
````

````python
# Block e
print(a)
````
