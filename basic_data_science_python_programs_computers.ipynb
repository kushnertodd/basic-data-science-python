# Computers
When computers were first invented, they had two components.
The first was a memory, in which data could be stored.
The second was electronics that implemented the calculation for which the computer was to be used.
The steps of the calculation were implemented by circuits that required a specific wiring to get the desired result.
When a different calculation was needed, the electronics were rewired to change the steps used in the calculation.

The rewiring was tedious and prone error. Early computer manufacturers had a better idea. They thought of putting the instructions for the calculations in the memory with the data.
The, a computer processing unit (CPU) would:
- read the instructions from the memory
- interpret what calculations the instructions intended to be done
- read data from the memory that was required for the calculation
- perform the calculation on the data
- write the calculation result back to the memory

This relieved the need to rewire the computer every time the calculation changed.
It did, however, create the need to design and interpret instructions that could carry out any calculation desired. These instructions became the first *programs*.




## Instructions
Instructions required three parts:
- *reference* in the memory where data to be read for the calculation was stored.
- a code for the calculation to carry out.
- a reference in the memory in which to put the result of the calculation.

The code could indicate every calculation that would be necessary.
These included addition, subtraction, sometimes just moving data from one place in memory to another, and the like.
The reference would indicate where in the memory a unit of data would reside, where each memory location was assigned a number.
The instruction would look something like:
- instruction code
- input data memory reference(s), possibly two in the case of addition
- output data memory reference

Instructions, being in memory, were also at memory locations that were assigned a number.
Instructions were by and large performed sequentially, with the CPU reading reading instructions from successive memory locations.
Some instructions could cause the CPU to read the next instruction from a different memory location, such as in the case of a conditional test ("if this is true perform this else perform that").
Instructions would be performed until a last instruction would indicate the end of the program.

The data in memory was being changed as the instructions were processed.
The set of values in memory at any point in time was the *data state*.
The first step in the program was to populate the memory with data, then perform changes on the data, then finally make changes externally visible based on the data such as creating displays or saving the data.
The program would behave conditionally based on the state of the data initially and as the data changed.


## High-level Languages
Memory could only store data as numbers.
Originally instructions, being stored in memory with data, were also stored as numbers.
Writing instructions as numbers turned out to be a tedious process.
As one of the originally programmers of computers, [Grace Hopper](https://president.yale.edu/biography-grace-murray-hopper), as a professor at Vassar and an officer in the Navy, had a better idea.
She first thought of using text-based programs instead of numbers.
She created COBOL as a simplified programming language using English words representing instructions.

These programs were stored in memory and a *compiler* would convert them to numbers that were the instructions for the CPU to perform.
From this language, eventually modern programming languages such as [Python](https://pythoninstitute.org/about-python) would be derived.

While these programming languages allow programmers not to think of programs as numeric instructions, like an automatic transmission allows a driver not to think in terms of changing of gears, there are times it is useful to understand what a program is doing "under the hood" to know why a program is behaving as it is and ensure it is performing as you want it to.
