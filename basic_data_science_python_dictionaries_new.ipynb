## Dictionaries
Dictionaries are groups of entries where each entry has a *key* that is used to look up a value, like looking up a word definition in a real dictionary.
Dictionary entries are unordered.
Dictionary keys can be any basic type and values are of any type.

### `dict()` function
There are three ways to use `dict()`:

| function | description |
| - | - |
| `dict(`*key*`=`*value*`, `*key*`=`*value*`, ...)` | create with a list of *key*`=`*value* pairs |
| `dict((`*key*`, `*value*`), (`*key*`, `*value*`), ...)` | create with a list of `(`*key*`, `*value*`)` pairs|
| `dict(zip(`*key-list*`, `*value-list*`))` | using the `zip()` to pair entries from a *key-list* and a *value-list* |

#### `dict(`*key*`=`*value*`, `*key*`=`*value*`, ...)`
Create a dictionary with a list of *key*`=`*value* pairs.

````python
print(dict(Frodo="Hobbit", Gandalf="Wizard", Legolas="Elf", Gimli="Dwarf"))
````

#### `dict((`*key*`, `*value*`), (`*key*`, `*value*`), ...)`
Create a dictionary with a list of `(`*key*`, `*value*`)` pairs.

````python
print(dict([("Frodo", "Hobbit"), ("Gandalf", "Wizard"), ("Legolas", "Elf"), ("Gimli", "Dwarf")]))
````

#### `dict(zip(`*key-list*`, `*value-list*`))`
Create a dictionary using the `zip()` function to pair entries from a *key-list* and a *value-list*.

````python
print(dict(zip(["Frodo", "Gandalf", "Legolas", "Gimli"], ["Hobbit", "Wizard", "Elf", "Dwarf"])))
````

## Accessing entries
Dictionary entry values can be accessed by key using square brackets `[]`.
There are two ways to access dictionary values.

| method | description |
| - | - |
| `[`*key*`]` | the value associated with *key* |
| `get(`*key*`)` | the value associated with *key*  |

````python
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}["Frodo"])
````

## Searching dictionaries
There are two operators to find items in a dictionary.

| method | description |
| - | - |
| [`in`](#scrollTo=in_operator) | whether an key is in a dictionary |
| [`not in`](#scrollTo=not_in_operator) | whether an key isn't in a dictionary |

### in operator
Whether an key is in a dictionary.

````python
# test if a key is in dictionary
print("Frodo" in {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"})
````

## Modifying dictionaries
The `list(`*dictionary*`)` function creates a list of the dictionary keys.
There are two functions that change the contents of a dictionary.

| method | description |
| - | - |
|	`list(`*dictionary*`)` | a list of the dictionary keys |
| `list(`*dictionary*`.keys())` | a list of the dictionary keys |
| `list(`*dictionary*`.values())` | a list of the dictionary values |

````python
print(list({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}))
````

- __Exercise__. Print the keys of `{"Frodo": "Hobbit", "Gandalf": "Wizard"}

- __Exercise__. Print the values of `{"Frodo": "Hobbit", "Gandalf": "Wizard"}`.

````python
# Set d[key] to value
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d["Gandalf"] = "Mithrandir"
print(d)
````

````python
# Remove d[key] from d. Raises a KeyError if key is not in the map
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
del d["Gandalf"]
print(d)
````

````python
# Remove all items from the dictionary
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d.clear()
print(d)
````

````python
# Return a shallow copy of the dictionary
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.copy())
````

````python
# Return a new view of the dictionary’s items ((key, value) pairs)
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(list(d.items()))
````

````python
# If key is in the dictionary, remove it and return its value, else return default.
# If default is not given and key is not in the dictionary, a KeyError is raised.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.pop("Gandalf"))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.pop("Gandalf"))
print(d)
````

````python
# If key is in the dictionary, remove it and return its value, else return default.
# If default is not given and key is not in the dictionary, a KeyError is raised.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.pop("missing", "not there"))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.pop("missing", "not there"))
print(d)
````

````python
# Remove and return a (key, value) pair from the dictionary. Pairs are returned in LIFO order.
# popitem() is useful to destructively iterate over a dictionary, as often used in set algorithms.
# If the dictionary is empty, calling popitem() raises a KeyError.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}.popitem())
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.popitem())
print(d)
print(d.popitem())
print(d)
````

````python
# Return a reverse iterator over the keys of the dictionary. This is a shortcut for reversed(d.keys())
print(list(reversed({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"})))
````

````python
# Update the dictionary with the key/value pairs from mapping or iterable and kwargs, overwriting existing keys. Return None.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"}.update({"Gimli": "Dwarf"}))
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
print(d.update({"Gimli": "Dwarf"}))
print(d)
````

````python
# Create a new dictionary with the merged keys and values of d and other, which must both be dictionaries.
# The values of other take priority when d and other share keys.
print({"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"} | {"Gimli": "Dwarf"})
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf"}
d |= {"Gimli": "Dwarf"}
print(d)
d = {"Frodo": "Hobbit", "Gandalf": "Wizard", "Legolas": "Elf", "Gimli": "Dwarf"}
d |= {"Gandalf": "Mithrandir"}
print(d)
````
