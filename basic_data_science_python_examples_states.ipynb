## Process a comma separated US States properties file
Here is a Python program to get data for US States (name, abbreviation, population, number of representatives, etc.) and sum the states' properties for each region and for all states.

The data will be read from the internet, given that is a convenient way to read a file in Google Colab.
The `requests` library will read raw data from a web link.
Here a list of raw state data on each comma separated line of the file:

| field | description |
| - | - |
| name | state name |
| abbrev | state name abbreviation |
| admitted | when the state was admitted to the US |
| population 2020 | state population in 2020 |
| sq miles | state size in square miles |
| sq km | state size in square kilometers |
| reps | number of state congressional representatives |

Read in the state properties comma separated file.
Each line has the format:

> *name*`,`*abbrev*`,`*admitted*`,`*population 2020*`,`*sq miles*`,`*sq km*`,`*reps*

Read the comma separated states' properties file `states.csv`.

Note that the code blocks depend on having run all previous code blocks.
Run the code blocks in order.

```python
import requests

url = "https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/refs/heads/main/data/States/states.csv"
response = requests.get(url)
data = response.text
```

Split the states properties file data into a list of each state properties line.

```python
states_properties_lines = data.splitlines()
print("Raw state properties lines:")
for i in range(5):
    print("   ", states_properties_lines[i])
```

Convert each line into a list of the state and its properties:

> `['`*name*`', '`*abbrev*'`, '`*admitted*`', '`*population 2020*`', '`*sq miles*`', '`*sq km*`', '`*reps*`']`

Then create a list of each of those state properties lists:
> `[[`*states-properties*`], [`*states-properties*`], ..., [`*states-properties*`]]`

We skip the first line with the field names, it's not needed.
This is a common pattern:
- Set a boolean variable to True.
- In a loop, first check the boolean variable.
- If it's true, set it to false and do nothing that first time.
- If it's not, continue with the business of the loop .

```python
states = []
header = True
for state_property_line in states_properties_lines:
    if header:
      # skip header line
      header = False
    else:
      state_properties = state_property_line.split(',')
      states.append(state_properties)
print("State properties lines split into lists:")
for i in range(5):
    print("   ", states[i])
```

Define some data structures to find the regions for states.

```python
# list of state regions
regions = [
    "New England",
    "Southwest",
    "Mid-Atlantic",
    "Southeast",
    "Midwest",
    "West"
]

# dictionary to convert a state to its region
states_regions = {
    "Connecticut": "New England",
    "Maine": "New England",
    "Massachusetts": "New England",
    "New Hampshire": "New England",
    "Rhode Island": "New England",
    "Vermont": "New England",
    "Arizona": "Southwest",
    "New Mexico": "Southwest",
    "Oklahoma": "Southwest",
    "Texas": "Southwest",
    "Delaware": "Mid-Atlantic",
    "Maryland": "Mid-Atlantic",
    "New Jersey": "Mid-Atlantic",
    "New York": "Mid-Atlantic",
    "Pennsylvania": "Mid-Atlantic",
    "Alabama": "Southeast",
    "Arkansas": "Southeast",
    "Florida": "Southeast",
    "Georgia": "Southeast",
    "Kentucky": "Southeast",
    "Louisiana": "Southeast",
    "Mississippi": "Southeast",
    "North Carolina": "Southeast",
    "South Carolina": "Southeast",
    "Tennessee": "Southeast",
    "Virginia": "Southeast",
    "West Virginia": "Southeast",
    "Illinois": "Midwest",
    "Indiana": "Midwest",
    "Iowa": "Midwest",
    "Kansas": "Midwest",
    "Michigan": "Midwest",
    "Minnesota": "Midwest",
    "Missouri": "Midwest",
    "Nebraska": "Midwest",
    "North Dakota": "Midwest",
    "Ohio": "Midwest",
    "South Dakota": "Midwest",
    "Wisconsin": "Midwest",
    "Alaska": "West",
    "California": "West",
    "Colorado": "West",
    "Hawaii": "West",
    "Idaho": "West",
    "Montana": "West",
    "Nevada": "West",
    "Oregon": "West",
    "Utah": "West",
    "Washington": "West",
    "Wyoming": "West",
}
```

For each region, total each property for the states in the region.
Then also create an overall total for each state's properties.

```python
regions_property_totals = {
    "New England":  {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Southwest":    {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Mid-Atlantic": {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Southeast":    {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Midwest":      {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "West":         {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
}
overall_property_totals = {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0};
header = True
for state in states:
    state_fields = state
    # get the state's properties
    state = state_fields[0]
    population_2020 = int(state_fields[3])
    sq_miles = int(state_fields[4])
    sq_km = int(state_fields[5])
    rep = int(state_fields[6])
    region = states_regions[state]
    # compute region property totals
    region_property_totals = regions_property_totals[region]
    region_property_totals["population_2020"] += population_2020
    region_property_totals["sq_miles"] += sq_miles
    region_property_totals["sq_km"] += sq_km
    region_property_totals["reps"] += rep
    # compute overall property totals
    overall_property_totals["population_2020"] += population_2020
    overall_property_totals["sq_miles"] += sq_miles
    overall_property_totals["sq_km"] += sq_km
    overall_property_totals["reps"] += rep
```

Print out the final property totals for each region and overall.

```python
print("region          population 2020   sq_miles      sq_km    reps")
print("------------    ---------------  ----------  ----------  ----")
for region in regions:
  region_property_totals = regions_property_totals[region]
  print(f"{region:<15} {region_property_totals["population_2020"]:>15,} {region_property_totals["sq_miles"]:>11,} {region_property_totals["sq_km"]:>11,} {region_property_totals["reps"]:>5}")
print("------------    ---------------  ----------  ----------  ----")
print(f"totals          {overall_property_totals["population_2020"]:>15,} {overall_property_totals["sq_miles"]:>11,} {overall_property_totals["sq_km"]:>11,} {overall_property_totals["reps"]:>5}")
```
