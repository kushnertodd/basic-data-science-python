## A simple Python program to read in data and prepare it for processing
Here is a Python program to get data for US States (name, abbreviation, population, number of representatives, etc.) as a first step for data analysis.

The data will be read from the internet, given there is no convenient way to read it from a file in Google Colab.
The `requests` library will read raw data from a web link.
Here a list of raw state data is read with each line having the format:

| field | description |
| - | - |
| name | state name |
| abbrev | state name abbreviation |
| admitted | when the state was admitted to the US |
| population 2020 | state population in 2020 |
| sq miles | state size in square miles |
| sq km | state size in square kilometers |
| reps | number of state congressional representatives |

Note that the code blocks depend on having run all previous code blocks.
Run the code blocks in order.

Read in the state properties comma separated file.
Each line has the format:

> *name*`,`*abbrev*`,`*admitted*`,`*population 2020*`,`*sq miles*`,`*sq km*`,`*reps*





Read the comma separated state properties file `states.csv`.

```python
import requests

url = "https://raw.githubusercontent.com/kushnertodd/basic-data-science-python/refs/heads/main/data/States/states.csv"
response = requests.get(url)
data = response.text
```

Split the states properties file data into a list of each state properties line.

```python
states_properties_lines = data.splitlines()
print("Raw state properties lines:")
for i in range(5):
    print("   ", states_properties_lines[i])
```

Convert each line into a list of the state and its properties.

> `['`*name*`', '`*abbrev*'`, '`*admitted*`', '`*population 2020*`', '`*sq miles*`', '`*sq km*`', '`*reps*`']`

Create a list of each state's properties lists.
> `[[`*states-properties*`], [`*states-properties*`], ..., [`*states-properties*`]]`

We skip the first line with the field names, it's not needed.

```python
states = []
header = True
for state_property_line in states_properties_lines:
    if header:
      # skip header line
      header = False
    else:
      state_properties = state_property_line.split(',')
      states.append(state_properties)
print("State properties lines split into lists:")
for i in range(5):
    print("   ", states[i])
```

Define some data structures to find the regions for states.

```python
# national state regions
regions = [
    "New England",
    "Southwest",
    "Mid-Atlantic",
    "Southeast",
    "Midwest",
    "West"
]

# a dictionary for the states' regions
states_regions = {
    "Connecticut": "New England",
    "Maine": "New England",
    "Massachusetts": "New England",
    "New Hampshire": "New England",
    "Rhode Island": "New England",
    "Vermont": "New England",
    "Arizona": "Southwest",
    "New Mexico": "Southwest",
    "Oklahoma": "Southwest",
    "Texas": "Southwest",
    "Delaware": "Mid-Atlantic",
    "Maryland": "Mid-Atlantic",
    "New Jersey": "Mid-Atlantic",
    "New York": "Mid-Atlantic",
    "Pennsylvania": "Mid-Atlantic",
    "Alabama": "Southeast",
    "Arkansas": "Southeast",
    "Florida": "Southeast",
    "Georgia": "Southeast",
    "Kentucky": "Southeast",
    "Louisiana": "Southeast",
    "Mississippi": "Southeast",
    "North Carolina": "Southeast",
    "South Carolina": "Southeast",
    "Tennessee": "Southeast",
    "Virginia": "Southeast",
    "West Virginia": "Southeast",
    "Illinois": "Midwest",
    "Indiana": "Midwest",
    "Iowa": "Midwest",
    "Kansas": "Midwest",
    "Michigan": "Midwest",
    "Minnesota": "Midwest",
    "Missouri": "Midwest",
    "Nebraska": "Midwest",
    "North Dakota": "Midwest",
    "Ohio": "Midwest",
    "South Dakota": "Midwest",
    "Wisconsin": "Midwest",
    "Alaska": "West",
    "California": "West",
    "Colorado": "West",
    "Hawaii": "West",
    "Idaho": "West",
    "Montana": "West",
    "Nevada": "West",
    "Oregon": "West",
    "Utah": "West",
    "Washington": "West",
    "Wyoming": "West",
}
```

This splits the data lines into lists making each field available individually.

```python
region_statistics = {
    "New England":  {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Southwest":    {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Mid-Atlantic": {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Southeast":    {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "Midwest":      {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
    "West":         {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0},
}
total_statistics = {"population_2020": 0, "sq_miles": 0, "sq_km": 0, "reps": 0};
header = True
for state in states:
    state_fields = state
    # get state statistics
    state = state_fields[0]
    population_2020 = int(state_fields[3])
    sq_miles = int(state_fields[4])
    sq_km = int(state_fields[5])
    rep = int(state_fields[6])
    region = states_regions[state]
    # compute region statistics
    statistics = region_statistics[region]
    statistics["population_2020"] += population_2020
    statistics["sq_miles"] += sq_miles
    statistics["sq_km"] += sq_km
    statistics["reps"] += rep
    # compute total statistics
    total_statistics["population_2020"] += population_2020
    total_statistics["sq_miles"] += sq_miles
    total_statistics["sq_km"] += sq_km
    total_statistics["reps"] += rep
```

Print out the final totals

```python
print("region          population 2020   sq_miles      sq_km    reps")
print("------------    ---------------  ----------  ----------  ----")
for region in regions:
  statistics = region_statistics[region]
  print(f"{region:<15} {statistics["population_2020"]:>15,} {statistics["sq_miles"]:>11,} {statistics["sq_km"]:>11,} {statistics["reps"]:>5}")
print("------------    ---------------  ----------  ----------  ----")
print(f"totals          {total_statistics["population_2020"]:>15,} {total_statistics["sq_miles"]:>11,} {total_statistics["sq_km"]:>11,} {total_statistics["reps"]:>5}")
```
