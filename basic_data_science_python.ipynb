### Escape Sequences
A backslash `\` in a string "escapes" the normal way it reads and interprets the next character, and gives it a new meaning as a special character or to include formatting in a string.

These are the most common backslash escapes:

| Escape sequence | Name | Use |
| - | - | - |
| \n | newline (line break) | break text across multiple lines |
| \t | tab (indent) | indent text to a fixed next tab stop |
| \\' | single quote | Insert a single quote in a single quoted string |
| \” | double quote | Insert a double quote in a double quoted string |
| \\\ | backslash |  Insert a backslash in a single or double quoted string |

#### String Slices
String *slices* accessed sequences of characters by referencing positions in square brackets separated by one or more colons `:`.
The ways to specify sequences are:
- *string*`[`*start*`:`*end* `]`
- *string*`[`*start*`:`*end* `:`*step*`]`

The characters are selected from the start position up to, but not including, the end position.
If a step is specified, that many characters are skipped for each character selected bwtween the start and end positions.

````python
# the first word
print("DON'T PANIC!"[0:6])

# the second word
print("DON'T PANIC!"[6:12])
````

Either the first or second position may be omitted.
- If the first position is omitted, the slice starts at the beginning of the string, up to but not including the second position.
- If the second position is omitted, the slice starts at the first character and ends with the last character of the string.

````python
# the first word with the first position omitted
print("DON'T PANIC!"[:5])

# the second word with the second position omitted
print("DON'T PANIC!"[7:])
````

### Searching Strings
Searching strings includes the following functions:

| Function | Description |
| - | - |
| `find()` | Find the starting position of the first occurrence of a substring |
| `rfind()` | Find the starting position of the last occurrence of a substring |


#### `find()`
There are three forms of `find()`:

| form | meaning |
| - | - |
| `find(`*substring*`)` | Find first position of *substring* |
| `find(`*substring*`, `*start*`)` |  Find first position of *substring* beginning at *start* |
| `find(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* between *start* and *end* |

`find()` returns position -1 if *substring* is not found.

````python
# find the position of first PANIC
#      0         1         2         3
#      01234567890123456789012345678901234567
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC"))
# find the position of second PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 14))
# find the position of PANIC between positions 14 and 26
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 14, 26))
# the position of PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 35))
````

#### `rfind()`
There are three forms of `rfind()`:

| form | meaning |
| - | - |
| `rfind(`*substring*`)` | Find first position of *substring* from the end|
| `rfind(`*substring*`, `*start*`)` |  Find first position of *substring* from the end beginning at *start* |
| `rfind(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* from the end between *start* and *end* |

`rfind()` returns position -1 if *substring* is not found.

````python
# find the position of rightmost PANIC
#      0         1         2         3
#      01234567890123456789012345678901234567
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC"))
# find the position of rightmost starting at position 14 PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 14))
# find the position of rightmost PANIC between positions 14 and 26
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 14, 26))
# the position of rightmost PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 35))
````

### Formatting Strings
Strings may be formatted with the following operations:

| Operation | Description |
| - | - |
| `center()` | The string centered in a string of a given width |
| `ljust()` | The string left justified in a string of a given width |
| `rjust()` | The string right justified in a string of a given width |

#### `center()`
There are two forms of `center()`:

| form | meaning |
| - | - |
| `center(*width*)` | The string centered in a blank string of a given width |
| `center(`*width*`, `*fillchar*`)` |  The string centered in a string of a given width filled out with *fillchar* |

````python
print("DON'T PANIC!".center(20))
print("DON'T PANIC!".center(20, '-'))
````

````python
print("DON'T PANIC!".ljust(20))
print("DON'T PANIC!".ljust(20, '-'))
````

`rjust()`

### Modifying Strings
Strings may be modified with the following operations:

| Operation | Description |
| - | - |
| `removeprefix()` | The string with a leading substring removed |
| `removesuffix()` | The string with a trailing substring removed |
| `replace()` | The string with a substring replaced a given of times |
| `join()` | The string with characters separated with a given substring |

````python

# strings can have a fixed leading substring removed
print("So Long, and Thanks for All the Fish".removeprefix('So Long, and '))

# strings can have a fixed ending substring removed
print("So Long, and Thanks for All the Fish".removesuffix(', and Thanks for All the Fish'))

# all string substrings can be replaced
print("row, row, row your boat".replace('row', 'paddle'))

# a fixed number of string substrings can be replaced
print("row, row, row your boat".replace('row', 'paddle', 2))

# join string characters separated by a fill character
print("-".join("DON'T PANIC!"))
````

## Integers
Integers are negative and positive whole numbers, without fractions.
The result of any integer operation is always an integer.


````python
# use '+' for integer addition
print(17 + 25)
# use '-' for integer subtraction
print(66 - 24)
# use '*' for integer multiplication
print(2 * 21)
# '//' performs integer division (integer result, remove any fraction)
print(128 // 3)
# use '%' for integer modulo (remainder after division)
print(128 % 43)
# integer absolute value
print(abs(-42))
# integer exponentiation
print(42 ** 2)
# convert an integer to a string
print(str(42))
# convert a string to an integer
print(int("42"))
````

## Special Sequences
A special sequence is a backslash `\` followed by one of the characters in the list below, where the backslash *escapes* the following character and has a special meaning.
In match patterns, in a single or double quoted string the backslash must be escaped with another backslash.
Simpler is to use a raw string `r'...'` or 'r"..."` that allow a bare backslash as described in the Strings section.

| Character | Description |
| - | - |
| \A | Returns a match if the specified characters are at the beginning of the string |
| \b | Returns a match where the specified characters are at the beginning or at the end of a word |
| \B | Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word |
| \d | Returns a match where the string contains digits (numbers from 0-9) |
| \D | Returns a match where the string DOES NOT contain digits |
| \s | Returns a match where the string contains a white space character |
| \S | Returns a match where the string DOES NOT contain a white space character |
| \w | Returns a match where the string contains any word characters (characters from a to Z, digits from 0-9, and the underscore _ character) |
| \W | Returns a match where the string DOES NOT contain any word characters |
| \Z | Returns a match if the specified characters are at the end of the string |


## Sets
A set is a set of characters inside a pair of square brackets [] with a special meaning. Escape a square bracket to include an actual square bracket in a pattern, `\[` or `\]`.

| Set | Description |
| - | - |
| [arn] | Returns a match where one of the specified characters (**a**, **r**, or **n**) is present |
| [a-n] | Returns a match for any lower case character, alphabetically between **a** and **n** |
| [^arn] | Returns a match for any character EXCEPT **a**, **r**, and **n** |
| [0123] | Returns a match where any of the specified digits (**0**, **1**, **2**, or **3**) are present |
| [0-9] | Returns a match for any digit between **0** and **9** |
| [0-5][0-9] | Returns a match for any two-digit numbers from **00** and **59** |
| [a-zA-Z] | Returns a match for any character alphabetically between **a** and **z**, lower case OR upper case |
| [+] | In sets, **+**, *****, **.**, **\|**, **()**, **$**, **{}** has no special meaning, so **[+]** means: return a match for any **+** character in the string |

# Placeholders
A dataset is the collection of the data being processed in a data analytics session.
The current values for all that data at any point in time is a *data state*.
The initial data state are all the available raw data.
This data is modified step by step as it is cleaned, analyzed, transformed, and visualized.

Placeholders, or *variables*, are needed to hold the data from step to step.
The placeholders are called *variables* because the data they hold may be changed from step to step.
This is distinct from mathematical variables, whose value never changes once it is associated with them.
The data state are the values contained by all existing variables.

When all steps are completed, the data is no longer needed and is released.
Any data needed for later must be saved before it is removed.

# Programs
A program automates the processing of data through data states, from the initial raw data until the last data state is handled and the session is over.
A program is a recipe and consists of a series of steps where some data is needed for each step, the data is modified, and the changed data results in a new data state. A recipe for making bread gives an example.

- initial data state: (flour, water, and yeast)
- make the dough
  - input: flour, water, and yeast
  - processing: mix and let rise
  - output: raw dough
  - data state: (raw dough)
- knead the dough
  - input: raw dough
  - processing: roll, press, and fold the dough
  - output: smooth dough
  - data state: (smooth dough)
- bake the dough:
  - input: smooth dough
  - processing: heat in the oven
  - output: bread
  - data state: (bread)

The initial data state is (flour, water, yeast) and the final data state is (bread).

In Python, the data state is stored in variables as the types of data and data structures above.
At each step, the data held by a variable may change,
or a variable may be assigned new data values, in which case it no longer needs the previous value and it is released (called *garbage collection*).
The data state is then the new set of values assigned to all variables.

## Program Statements
A program statement is responsible for each step of the processing that changes the data state.
Program statements may:
- assign data to a variable
- modify the data associated with a variable
- pass data to a function that uses the data to cause some external effect such as:
  - saving the data
  - creating a report from the data
  - visualizing the data
  - giving the data to a machine learning algorithm to predict future effects

Each statement modifies the data state.
The progression of a program as it is run is:
```
(initial data state)
program statement 1
(data state 1)
program statement 2
(data state 2)
...
last program statement
(all data is released and the program ends)
```

## Statements that control statement processing order
The most powerful feature of programs is that they can process huge amounts of data and automatically make decisions based on data values.
Statements are normally processed from the top of the program to the end.
There are two statements that select whether a group of statements are to be processed:
- `if` statements that select whether to process a group of statements based on a condition.
- `match` statements that select a group of statements to process based on a value.

There are also two statements that allow processing groups of statements repeatedly:
- `while` statements repeat a group of statements while a condition is true.
- `for` statements that process a group of statements once for each data value in a list.

The statements being controlled by these are grouped by being indented with the exact same indentation for each statement as a *block*.
These statements in each block are then processed from first to last.

### `if` Statements
A block of statements are processed together if a condition is true with an `if` statement.
The condition follows the `if` and a colon `:` ends the line.

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
````

One variation of an `if` statement allows one block of statements to be processed if the condition is true and another to be processed if the condition is false.
The statements to be processed if the condition are false follow an `else` statement that matches an `if` statement that has the same size indentation.
The `else` is followed by a colon `:`.

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
else:
    print('That is not quite the answer we were looking for')
````

Another variation of an `if` statement allows multiple conditions to be tested one by one and process different groups of statements depending on the resuts.If there are multiple conditions to be tested, the first condition is with an `if` statement at the beginning, then rest of the conditions are with an `elif` statement.
The `elif` statement is followed by a colon `:`.
The statements following the `elif` have the same size indentation.
There can be a `else` statement after the last `elif` statement,
though that is not required

````python
x = 42
if x == 42:
    print('We have the Answer to the Ultimate Question of Life, the Universe, and Everything')
elif x == 41 or x == 43:
    print('Close, but not quite')
else:
    print('That is not quite the answer we were looking for')
````

There is one more form of `if` statement called a "ternary if".
It is a form of `if` statement that returns a value and is used like this:

*true_value* `if` *condition* `else` *false_value*

where:
- *true_value* is the value if the condition is `True`
- *false_value* is the value if the condition is `False`

````python
the_answer = 42
result = "correct" if the_answer == 42 else "incorrect"
print(result)
````

### `match` Statements
A match statement takes a value and processes the `case` blocks matching the value.
This works if the condition is that an expression is equal to a single value.

````python
word = "abc"
n = len(word)
match n:
    case 1:
        print(word, " has 1 character")
    case 2:
        print(word, " has 2 characters")
    case 3:
        print(word, " has 3 characters")
````

This is like an `elif` statement but is clearer and easier to read.

````python
word = "abc"
n = len(word)
if n == 1:
        print(word, " has 1 character")
elif n == 2:
        print(word, " has 2 characters")
elif n == 3:
        print(word, " has 3 characters")
````

There can be a the last `case` block with the value `_` that acts like an `else` if no other block matches.

````python
word = "abc"
n = len(word)
match n:
    case 1:
        print(word, " has 1 character")
    case 2:
        print(word, " has 2 characters")
    case 3:
        print(word, " has 3 characters")
    case _:
        print(word, " has", n, "characters")
````

### `while` Statements
Statements can be processed as long as a condition is true with the `while` statement.

````python
# print all even numbers up to 10
n = 2
while (n <= 10):
  print(n)
  n += 2
````

In the version `range(`*start*`, `*finish*`, `*step*`)`,
the last number will be less than *finish-1* if the last *step* takes it past that value.

````python
# prints numbers from  0 to 4
for i in range(5):
    print(i)
````

### break and continue Statements
The break statement breaks out of the innermost enclosing for or while loop:

````python
# print squares of numbers until one is divisible by 3
for i in range(2, 10, 2):
  if i % 3 == 0:
     print(i, "is dividible by 3")
     break
  print(i)
````

````python
# print squares of numbers not divisible by 3
for i in range(2, 10, 2):
  if i % 3 == 0:
      print(i, "is dividible by 3, skipping")
      continue
  else:
      print(i*i)
````

## A Python program to perform data cleaning
Let us put together everything we've learned so far to perform a real data science function: data cleaning of a list of Audible audio books. The format of the data is:

| Column | Breakdown |
| - | - |
| name | Name of the audiobook |
| author | Author of the audiobook |
| narrator | Narrator of the audiobook |
| time | Length of the audiobook |
| releasedate | Release date of the audiobook |
| language | Language of the audiobook |
| stars | No. of stars the audiobook received |
| price | Price of the audiobook in INR |

There is a data cleaning task for a number of fields, which we will discover by looking at the raw data.
For each field that needs cleaning, we will make a decision of how to do so.

````python
import requests

uncleaned_url = "https://raw.githubusercontent.com/kushnertodd/basic-data-science/refs/heads/main/data/Audible%20Dataset/audible_uncleaned.tsv"
uncleaned_response = requests.get(uncleaned_url)
````

````python
'''
given the header fields are:
  ['name', 'author', 'narrator', 'time', 'releasedate', 'language', 'stars', 'price']
the field values for each record can be added to lists created with each of the fieldnames.
example:
  Geronimo Stilton #11 & #12,Writtenby:GeronimoStilton,Narratedby:BillLobely,2 hrs and 20 mins,04-08-08,English,5 out of 5 stars34 ratings,468.00
after splitting on ',' becomes
  ['Geronimo Stilton #11 & #12', 'Writtenby:GeronimoStilton', 'Narratedby:BillLobely', '2 hrs and 20 mins', '04-08-08', 'English', '5 out of 5 stars34 ratings', '468.00']
'''
header = True
uncleaned_names = []
uncleaned_authors = []
uncleaned_narrators = []
uncleaned_times = []
uncleaned_releasedates = []
uncleaned_languages = []
uncleaned_stars = []
uncleaned_prices = []
# Check if the request was successful
if uncleaned_response.status_code == 200:
    uncleaned_data = uncleaned_response.text
    for line in uncleaned_data.splitlines():
      if line.strip(): # line is not blank
        if header:     # skip header line
          header = False
        else:
          uncleaned_audible_fields = line.split('\t')
          uncleaned_names.append(uncleaned_audible_fields[0])
          uncleaned_authors.append(uncleaned_audible_fields[1])
          uncleaned_narrators.append(uncleaned_audible_fields[2])
          uncleaned_times.append(uncleaned_audible_fields[3])
          uncleaned_releasedates.append(uncleaned_audible_fields[4])
          uncleaned_languages.append(uncleaned_audible_fields[5])
          uncleaned_stars.append(uncleaned_audible_fields[6])
          uncleaned_prices.append(uncleaned_audible_fields[7])
    print('names        ', uncleaned_names [0:3])
    print('authors      ', uncleaned_authors [0:3])
    print('narrators    ', uncleaned_narrators [0:3])
    print('times        ', uncleaned_times [0:3])
    print('releasedates ', uncleaned_releasedates [0:3])
    print('languages    ', uncleaned_languages [0:3])
    print('stars        ', uncleaned_stars [0:3])
    print('prices       ', uncleaned_prices [0:3])
````

First it is worth determining all the different variant formats of the time field.
We can convert the digits of each field to 'd' to make the format generic:
```
d hrs and dd mins

```
We can then add the fields to a set which will contain a unique copy of each converted field.

````python
formatted_uncleaned_time_set = set()
first = 0
for uncleaned_time in uncleaned_times:
  # convert digits to 'd'
  formatted_uncleaned_time = re.sub(r'\d', 'd', uncleaned_time)
  formatted_uncleaned_time_set.add(formatted_uncleaned_time)
for formatted_uncleaned_time in formatted_uncleaned_time_set:
  print(formatted_uncleaned_time)
````

Now we will parse out each of the above formats into hours and minutes, and convert those into total minutes (60*hours + minutes).
They turn out to reduce to four cases ("[s]" implies an optional "s"):

- "Less than n minute" (n turns out to be '1')
- "*n* hr[s]" (minutes = 0)
- "*n* hr[s] and *n* min[s]"
- "*n* min[s]" (hours = 0)

The optional "s" ('[s]') turns out to match the regular expression `s?`.
We will introduce a custom function, `parse_duration_to_minutes(line)`, to parse out a single uncleaned time field and return the computed total minutes.

````python
import re

def parse_duration_to_minutes(uncleaned_time):
    uncleaned_time = uncleaned_time.strip()

    # 1. Handle special case: "Less than d minute" -> 1 minute
    if re.search(r'Less than \d minute', uncleaned_time):
        return 1

    # 2. Extract hours using 'hrs?' (matches 'hr' or 'hrs')
    hours_match = re.search(r'(\d+)\s*hrs?', uncleaned_time)
    hours = int(hours_match.group(1)) if hours_match else 0

    # 3. Extract minutes using 'mins?' (matches 'min' or 'mins')
    mins_match = re.search(r'(\d+)\s*mins?', uncleaned_time)
    minutes = int(mins_match.group(1)) if mins_match else 0

    # 4. Calculate total minutes
    return (hours * 60) + minutes

cleaned_times = []
for uncleaned_time in uncleaned_times:
  cleaned_time = parse_duration_to_minutes(uncleaned_time)
  cleaned_times.append(cleaned_time)
print('uncleaned_times ', uncleaned_times[0:6])
print('cleaned_times   ', cleaned_times[0:6])
````

For the release data fields of the format `6/11/2020` and `30-10-18` it is best to determine which of the fields are month, day, and year.

This is best done by determining the minimum and maximum for the three fields for each of these formats.
We can then convert them to one consistent format, say 'month/day/year`.

````python
import re

# parse all the dates like 6/11/2020
slashdate_field1_min = 99999
slashdate_field1_max = 0
slashdate_field2_min = 99999
slashdate_field2_max = 0
slashdate_field3_min = 99999
slashdate_field3_max = 0
for uncleaned_releasedate in uncleaned_releasedates:
   slash_match = re.search(r'(\d+)/(\d+)/(\d+)', uncleaned_releasedate)
   if slash_match:
    slashdate_field1 = int(slash_match.group(1))
    slashdate_field2 = int(slash_match.group(2))
    slashdate_field3 = int(slash_match.group(3))
    slashdate_field1_min = min(slashdate_field1_min, slashdate_field1)
    slashdate_field1_max = max(slashdate_field1_max, slashdate_field1)
    slashdate_field2_min = min(slashdate_field2_min, slashdate_field2)
    slashdate_field2_max = max(slashdate_field2_max, slashdate_field2)
    slashdate_field3_min = min(slashdate_field3_min, slashdate_field3)
    slashdate_field3_max = max(slashdate_field3_max, slashdate_field3)
print('slashdate_field1_min: ', slashdate_field1_min)
print('slashdate_field1_max: ', slashdate_field1_max)
print('slashdate_field2_min: ', slashdate_field2_min)
print('slashdate_field2_max: ', slashdate_field2_max)
print('slashdate_field3_min: ', slashdate_field3_min)
print('slashdate_field3_max: ', slashdate_field3_max)
````

Given field 1 and 2 ranges are both 1-12, we'll chalk this up to this being synthetic data and use our intuition that the dates are the usual US format "month/day/year".

Now repeat that for dates of the format '30-10-18'.

````python
import re

# parse all the dates like 30-10-18
hyphendate_field1_min = 99999
hyphendate_field1_max = 0
hyphendate_field2_min = 99999
hyphendate_field2_max = 0
hyphendate_field3_min = 99999
hyphendate_field3_max = 0
for uncleaned_releasedate in uncleaned_releasedates:
   hyphen_match = re.search(r'(\d+)-(\d+)-(\d+)', uncleaned_releasedate)
   if hyphen_match:
    hyphendate_field1 = int(hyphen_match.group(1))
    hyphendate_field2 = int(hyphen_match.group(2))
    hyphendate_field3 = int(hyphen_match.group(3))
    hyphendate_field1_min = min(hyphendate_field1_min, hyphendate_field1)
    hyphendate_field1_max = max(hyphendate_field1_max, hyphendate_field1)
    hyphendate_field2_min = min(hyphendate_field2_min, hyphendate_field2)
    hyphendate_field2_max = max(hyphendate_field2_max, hyphendate_field2)
    hyphendate_field3_min = min(hyphendate_field3_min, hyphendate_field3)
    hyphendate_field3_max = max(hyphendate_field3_max, hyphendate_field3)
print('hyphendate_field1_min: ', hyphendate_field1_min)
print('hyphendate_field1_max: ', hyphendate_field1_max)
print('hyphendate_field2_min: ', hyphendate_field2_min)
print('hyphendate_field2_max: ', hyphendate_field2_max)
print('hyphendate_field3_min: ', hyphendate_field3_min)
print('hyphendate_field3_max: ', hyphendate_field3_max)
````

It looks like those dates have the format 'day-month-year (2 digit)', where we assume the year is after 2000 (year = 2000 + "year (2 digit)").

Let's put this together.

````python
import re

cleaned_releasedates = []
for uncleaned_releasedate in uncleaned_releasedates:
  # detect dates like 6/11/2020, leave unchanged
  slash_match = re.search(r'(\d+)/(\d+)/(\d+)', uncleaned_releasedate)
  if slash_match:
    cleaned_releasedate = uncleaned_releasedate
    cleaned_releasedates.append(cleaned_releasedate)
  else:
    # detect dates like 30-10-18, convert to 10/30/2018
    hyphen_match = re.search(r'(\d+)-(\d+)-(\d+)', uncleaned_releasedate)
    if hyphen_match:
      hyphendate_field1 = int(hyphen_match.group(1))
      hyphendate_field2 = int(hyphen_match.group(2))
      hyphendate_field3 = int(hyphen_match.group(3))
      cleaned_releasedate = f'{hyphendate_field2}/{hyphendate_field1}/{2000+int(hyphendate_field3)}'
      cleaned_releasedates.append(cleaned_releasedate)
print('uncleaned_releasedates ', uncleaned_releasedates[0:10])
print('cleaned_releasedates ', cleaned_releasedates[0:10])
````
