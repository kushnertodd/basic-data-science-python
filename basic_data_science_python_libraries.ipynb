### Probabability functions
These are three math library functions related to probability:
| method | function | purpose |
| - | - | - |
| Factorial | `factorial(`*n*`)` | the number of ways to order n numbers |
| Permutations | `perm(`*n*`, `*k*`)` | the number of ways to order k items out of n items |

### Factorial
Factorial is the number of ways to order n numbers:
$$n!$$

````python
import math
# 5 factorial
print(math.factorial(5))
````

### Permutations
Permutations are the number of ways to order k items from n items:
$$ {^n\mathrm{P}_k} = \frac{n!}{(n-k)!} $$


````python
import math
# permutations of 3 items out of 7
print(math.perm(7, 3))
````

__Exercise__: compute ${^{10}\mathrm{P}_4}$.

### Combinations
Combinations are the number of ways to choose k items from n  
$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

````python
# combinations of choosing k items from n
import math
# combinations of 3 items out of 7
print(math.comb(7, 3))
````

## Special Sequences
A special sequence is a backslash `\` followed by one of the characters in the list below, where the backslash *escapes* the following character and has a special meaning.
In match patterns, in a single or double quoted string the backslash must be escaped with another backslash.
Simpler is to use a raw string `r'...'` or 'r"..."` that allow a bare backslash as described in the Strings section.

| Character | Description |
| - | - |
| \A | Returns a match if the specified characters are at the beginning of the string |
| \b | Returns a match where the specified characters are at the beginning or at the end of a word |
| \B | Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word |
| \d | Returns a match where the string contains digits (numbers from 0-9) |
| \D | Returns a match where the string DOES NOT contain digits |
| \s | Returns a match where the string contains a white space character |
| \S | Returns a match where the string DOES NOT contain a white space character |
| \w | Returns a match where the string contains any word characters (characters from a to Z, digits from 0-9, and the underscore _ character) |
| \W | Returns a match where the string DOES NOT contain any word characters |
| \Z | Returns a match if the specified characters are at the end of the string |


## Sets
A set is a set of characters inside a pair of square brackets [] with a special meaning. Escape a square bracket to include an actual square bracket in a pattern, `\[` or `\]`.

| Set | Description |
| - | - |
| [arn] | Returns a match where one of the specified characters (**a**, **r**, or **n**) is present |
| [a-n] | Returns a match for any lower case character, alphabetically between **a** and **n** |
| [^arn] | Returns a match for any character EXCEPT **a**, **r**, and **n** |
| [0123] | Returns a match where any of the specified digits (**0**, **1**, **2**, or **3**) are present |
| [0-9] | Returns a match for any digit between **0** and **9** |
| [0-5][0-9] | Returns a match for any two-digit numbers from **00** and **59** |
| [a-zA-Z] | Returns a match for any character alphabetically between **a** and **z**, lower case OR upper case |
| [+] | In sets, **+**, *****, **.**, **\|**, **()**, **$**, **{}** has no special meaning, so **[+]** means: return a match for any **+** character in the string |

### The findall() Function
The `findall() function returns a list containing all matches.

````python
# Print a list of all matches in the order they are found
import re
txt = "The rain in Spain"
re.findall(".ai.", txt)
````

````python
# Return an empty list if no match was found
import re
txt = "The rain in Spain"
re.findall("Portugal", txt)
````

### The search() Function
The `search()` function searches the string for a match.
- If there is more than one match, only the first occurrence of the match will be returned.

````python
# Search for the first white-space character in the string
# A raw string is used in the patternto include the backslash
import re
txt = "The rain in Spain"
re.search(r"\s", txt).start()
````

### The split() Function
The `split()` function returns a list where the string has been split at each match.

````python
# Split at each white-space character
# A raw string is used in the pattern to include the backslash
import re
txt = "The rain in Spain"
re.split(r"\s", txt)
````

You can control the number of occurrences by specifying the `maxsplit` parameter.

````python
# Split the string only at the first occurrence
import re

txt = "The rain in Spain"
re.split(r"\s", txt, maxsplit=1)
````

### The sub() Function
The `sub()` function replaces the matches with the text of your choice.

````python
# Replace every white-space character with the number 9
import re
txt = "The rain in Spain"
re.sub(r"\s", "-", txt)
````

You can control the number of replacements by specifying the `count` parameter.

````python
# Replace the first 2 occurrences
import re
txt = "The rain in Spain"
re.sub(r"\s", "-", txt, count=2)
````

### Matching with groups
Text matching a pattern can be captured into a group by putting the match pattern between parentheses `()`.
The text matching each group can be accessed separately.

````python
# match each field of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").groups()
````

````python
# get month of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").groups()[0]
````

````python
# get day of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").groups()[1]
````

````python
# get year of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").groups()[2]
````

````python
# alternate way to get match each field of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").group(0)
````

````python
# alternate way to get month of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").group(1)
````

````python
# alternate way to get day of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").group(2)
````

````python
# alternate way to get year of a date
import re
re.search(r"(\d{2})/(\d{2})/(\d{4})", "07/04/2022").group(3)
````

````python
# match each field of an email address
import re
re.search(r"(\w+)@(\w+)", "monty_python@gmail.com").groups()
````

````python
# match each field of a web address
import re
re.search(r"https://www.(\w+)\.(\w+)", "https://www.google.com").groups()
````

````python
# match each field of an address
import re
re.search(r"(\w+), (\d+ .+), (.*), (\w{2}) (\d+)", "Googleplex, 1600 Amphitheatre Pkwy, Mountain View, CA 94043").groups()
````
