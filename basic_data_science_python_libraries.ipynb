# Python Libraries

The best feature of Python are that it has open-source libraries for almost anything you want a program to do.
Some useful ones are:
- [Mathematical libraries](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_libraries_numerical.ipynb)
  - Mathematics (`math`)
  - Statistics (`statistics`)
- [Date/time](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_libraries_datetime.ipynb) (`datetime`)
- [Regular Expressions](https://colab.research.google.com/github/kushnertodd/basic-data-science-python/blob/main/basic_data_science_python_libraries_regex.ipynb) (`regex`)
- Data analysis (`pandas`)
- Numerical computing (`numpy`)
- Statistical and mathematical algorithms (`scipy`)
- Data visualization (`matplotlib`)
- Statistical graphics (`seaborn`)
- Machine learning (`scikit-learn`)

To use libraries, we use the `import` statement at the top of the program.
It makes the functions in the library available to Python.

`import` *library*
