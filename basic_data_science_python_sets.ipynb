## Sets
Sets are an unordered group of values with no duplicates.
Sets can contain values of any types.
Sets support operations defined by [set theory](https://realpython.com/python-sets/).
These are covered in the following topics.

| Topic | Description |
| - | - |
| Creating sets | Building sets from groups of items |
| Accessing items | Extracting items from sets |
| Searching sets | Finding substrings in sets |
| Set Properties | Obtaining sets information |
| Modifying sets | Changing set content |

## Creating Sets
Sets are constructed in a number of ways.

| method | description |
| - | - |
| braces | items between braces `{}` |
| set operators | operators on sets |
| set functions | functions that return sets |

```python
print({"apple", "banana", "cherry"})
print({"apple", "banana", "cherry", "apple"})
```

```python
print(len({"apple", "banana", "cherry"}))
```

```python
# A set can contain different data types:
print({"abc", 34, True, 40, "male"})
print({})
```

```python
print(set(["apple", "banana", "cherry"]))
print(set("cherry"))
print(set())
```

```python
print("banana" in {"apple", "banana", "cherry"})
print("pear" not in {"apple", "banana", "cherry"})
```

```python
print(len({"apple", "banana", "cherry"}))
```

```python
# Join set1 and set2 into a new set:
print({"a", "b", "c"}.union({1, 2, 3}))
print({"a", "b", "c"}.union({1, 2, 3}, {"apple", "bananas", "cherry"}))
```

```python
# Use | to join two sets:
print({"a", "b", "c"} | {1, 2, 3})
print({"a", "b", "c"} | {1, 2, 3} | {"apple", "bananas", "cherry"})
```

```python
print({"apple", "banana", "cherry"}.intersection({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} & {"google", "cherry", "apple"})
print({"apple", "banana", "cherry"} & {"google", "cherry", "apple"} & {"cherry", "pear"})
```

```python
print({"apple", "banana", "cherry"}.difference({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} - {"google", "microsoft", "apple"})
```

```python
print({"apple", "banana", "cherry"}.symmetric_difference({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} ^ {"google", "microsoft", "apple"})
```

| Method | Shortcut | Description |
| - | - | - |
| difference() | - | the items in one set but not in another |
| intersection() | & | the items in both of two sets  |
| isdisjoint() |   | whether two sets have no intersection |
| issubset() | <= | whether all items of one set are in another |
| issubset() | < | whether all items of one set are in another (and are not equal) |
| issuperset() | >=| whether all items of one set include all those in another |
| issuperset() | > | whether all items of one set include all those in another (and are not equal) |
| symmetric_difference() | ^ | the items not in both of two sets |
| union() | \| | all unique items on combining two sets |



```python
# set without duplicates
{'apple', 'orange', 'apple', 'pear', 'orange', 'banana'}
```

```python
# count the number of values in the set
len({'apple', 'banana', 'orange', 'pear'})
```

```python
# detect if a value is in a set
'orange' in {'apple', 'banana', 'orange', 'pear'}
```

```python
# detect if a value is not in a set
'grape' not in {'apple', 'banana', 'orange', 'pear'}
```

```python
# set intersection: fruit in both sets
{'apple', 'banana', 'orange', 'pear'} & {'grape', 'banana', 'peach', 'orange'}
```

```python
# set union: fruit in either set
{'apple', 'banana', 'orange', 'pear'} | {'grape', 'banana', 'peach', 'orange'}
```

```python
# set difference: fruit in one set but not in another
{'apple', 'banana', 'orange', 'pear'} - {'grape', 'banana', 'peach', 'orange'}
```

```python
# fruit in one or the othere set but not both
{'apple', 'banana', 'orange', 'pear'} ^ {'grape', 'banana', 'peach', 'orange'}
```

```python
# is one set a subset of another
{'banana', 'pear'} <= {'apple', 'banana', 'orange', 'pear'}
```

```python
# is one set a superset of another
{'apple', 'banana', 'orange', 'pear'} >= {'banana', 'pear'}
```
