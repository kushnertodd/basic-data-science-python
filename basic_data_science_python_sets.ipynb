Sets can contain values of any types.

## Creating Sets
Sets are constructed in two ways.

| method | description |
| - | - |
| braces | items between braces `{}` |
| `set()` function | converts items to sets |

### braces
Sets can be constructed by items between braces `{}`.
Sets eliminate duplicates.



```python
print({'apple', 'banana', 'cherry', 'apple'})
```

```python
print({})
```

### `set()` function
There are two ways to use `set()`:

| use | meaning |
| - | - |
| `set()` |  the empty set |
| `set(`*list*`)` |  a set of the items in *list* |
| `set(`*string*`)` | a set of the characters in *string* |

#### `set()` function
The empty set (displayed as "set()" instead of "{}").

```python
print(set())
```

#### `set(`*list*`)` function
Creates a set of the items in *list*. Duplicates are removed.


```python
print(set(['apple', 'banana', 'cherry', 'apple']))
```

#### `set(`*string*`)` function
Creates a set of the characters in *string*. Duplicates are removed.

```python
print(set('cherry'))
```

## Searching sets
There are two ways to find items in sets:

| use | meaning |
| - | - |
| `in` |  detect if a value is in a set |
| `not in` | detect if a value is not in a set |

### `in`
Detect if a value is in a set.

```python
'orange' in {'apple', 'banana', 'orange', 'pear'}
```

### `not in`
Detect if a value is not in a set.

```python
'grape' not in {'apple', 'banana', 'orange', 'pear'}
```

## Set Properties
There are eight ways to obtain set information.

| Method | Shortcut | Description |
| - | - | - |
| `<=` | whether all items of one set are in another |
| `<` | whether all items of one set are in another (and are not equal) |
| `>=` | whether all items of one set include all those in another |
| `>` | whether all items of one set include all those in another (and are not equal) |
| `isdisjoint()` | whether two sets have no intersection |
| `len()` | number of set items |
| `max()` | largest set item |
| `min()` | smallest set |


## `<=` operator
Whether all items of one set are in another.
A *subset*.



```python
{'banana', 'pear'} <= {'apple', 'banana', 'orange', 'pear'}
```

Whether all items of one set are in another (and are not equal). A *proper* subset.

## Set functions
Functions on sets.

| Function | Shortcut | Description |
| - | - | - |
| difference() | - | the items in one set but not in another |
| intersection() | & | the items in both of two sets  |
| symmetric_difference() | ^ | the items not in either of two sets |
| union() | \| | all unique items on combining two sets |

### *set*.`difference(`*set*`)` function (shortcut: `-`)
The items in one set but not in another.





```python
print({'apple', 'orange', 'cherry'}.difference({'cherry', 'pear', 'apple'}))
```

```python
print({'apple', 'orange', 'cherry'} - {'cherry', 'pear', 'apple'})
```

- __Exercise__:

### *set*.`intersection(`*set*`)` function (shortcut: `&`)
The items in both of two sets .


```python
print({'apple', 'orange', 'cherry'}.intersection({'cherry', 'pear', 'apple'}))
```

```python
print({'apple', 'orange', 'cherry'} & {'cherry', 'pear', 'apple'})
```

```python
print({'apple', 'orange', 'cherry'}.intersection({'cherry', 'pear', 'apple'}, {'cherry', 'grape'}))
```

```python
print({'apple', 'orange', 'cherry'} & {'cherry', 'pear', 'apple'} & {'cherry', 'grape'})
```

- __Exercise__:

### *set*.`symmetric_difference(`*set*`)` function (shortcut: `^`)
The items not in either of two sets.


```python
print({'apple', 'orange', 'cherry'}.symmetric_difference({'cherry', 'pear', 'apple'}))
```

```python
print({'apple', 'orange', 'cherry'} ^ {'cherry', 'pear', 'apple'})
```

- __Exercise__:

### *set*.`union(`*set*, *set*, `...)` function (shortcut: '|')
All unique items on combining sets.
Multiple sets can be combined.

```python
print({'apple', 'orange', 'cherry'}.union({'cherry', 'pear', 'apple'}))
```

```python
print({'apple', 'orange', 'cherry'} | {'cherry', 'pear', 'apple'})
```

```python
print({'apple', 'orange', 'cherry'}.union({'cherry', 'pear', 'apple'}, {'cherry', 'grape'}))
```

```python
print({'apple', 'orange', 'cherry'} | {'cherry', 'pear', 'apple'} | {'cherry', 'grape'})
```
