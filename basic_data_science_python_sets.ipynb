## Sets
Sets are an unordered group of values with no duplicates.
Sets can contain values of any types.
Sets support operations defined by [set theory](https://realpython.com/python-sets/).
These are covered in the following topics.

| Topic | Description |
| - | - |
| Creating sets | Building sets from groups of items |
| Searching sets | Finding substrings in sets |
| Set properties | Obtaining set information |
| Set functions | Functions on sets |

## Creating Sets
Sets are constructed in two ways.

| method | description |
| - | - |
| braces | items between braces `{}` |
| `set()` function | converts items to sets |

### braces
Sets can be constructed by items between braces `{}`.
Sets eliminate duplicates.



```python
print({"apple", "banana", "cherry", "apple"})
```

A set can contain any basic data types.

```python
print({"abc", 34, True, 40, "male"})
```

Sets can be empty.

```python
print({})
```

### `set()` function
There are two ways to use `set()`:

| use | meaning |
| - | - |
| `set(`*lists*`)` |  a set of the items in *list* |
| `set(`*string*`)` | a set of the characters in *string* |

#### `set(`*list*`)` function
Creates a set of the items in *list*.


```python
print(set(["apple", "banana", "cherry", "apple"]))
```

## Searching sets
There are two ways to find items in sets:

| use | meaning |
| - | - |
| `in` |  detect if a value is in a set |
| `not in` | detect if a value is not in a set |

### `in`
Detect if a value is in a set.

```python
"orange" in {"apple", "banana", "orange", "pear"}
```

### `not in`

## Set Properties
There are eight ways to obtain set information.

| Method | Shortcut | Description |
| - | - | - |
| `<=` | whether all items of one set are in another |
| `<` | whether all items of one set are in another (and are not equal) |
| `>=` | whether all items of one set include all those in another |
| `>` | whether all items of one set include all those in another (and are not equal) |
| `isdisjoint()` | whether two sets have no intersection |
| `len()` | number of set items |
| `max()` | largest set item |
| `min()` | smallest set |


## `<=` operator
Whether all items of one set are in another.



```python
{'banana', 'pear'} <= {'apple', 'banana', 'orange', 'pear'}
```

## `<` operator
Whether all items of one set are in another (and are not equal).


```python
{'banana', 'pear'} < {'apple', 'banana', 'orange', 'pear'}
```

## `>=` operator
Whether all items of one set include all those in another.


```python
{'apple', 'banana', 'orange', 'pear'} >= {'banana', 'pear'}
```

## Set functions
Functions on sets.

| Function | Shortcut | Description |
| - | - | - |
| difference() | - | the items in one set but not in another |
| intersection() | & | the items in both of two sets  |
| symmetric_difference() | ^ | the items not in both of two sets |
| union() | \| | all unique items on combining two sets |

### *set*.`difference()` function (shortcut: `-`)
The items in one set but not in another.





```python
print({"apple", "banana", "cherry"}.difference({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} - {"google", "microsoft", "apple"})
```

```python
{'apple', 'banana', 'orange', 'pear'} - {'grape', 'banana', 'peach', 'orange'}
```

### *set*.`intersection()` function (shortcut: `&`)
The items in both of two sets .


```python
print({"apple", "banana", "cherry"}.intersection({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} & {"google", "cherry", "apple"})
print({"apple", "banana", "cherry"} & {"google", "cherry", "apple"} & {"cherry", "pear"})
```

```python
{'apple', 'banana', 'orange', 'pear'} & {'grape', 'banana', 'peach', 'orange'}
```

### *set*.`symmetric_difference()` function (shortcut: `^`)
The items not in both of two sets.


```python
print({"apple", "banana", "cherry"}.symmetric_difference({"google", "microsoft", "apple"}))
```

```python
print({"apple", "banana", "cherry"} ^ {"google", "microsoft", "apple"})
```

```python
{'apple', 'banana', 'orange', 'pear'} ^ {'grape', 'banana', 'peach', 'orange'}
```

### *set*.`union(`*set*, *set*, `...)` function (shortcut: '|')
All unique items on combining sets.
Multiple sets can be combined.

```python
print({"a", "b", "c"}.union({1, 2, 3}))
```

```python
print({"a", "b", "c"}.union({1, 2, 3}, {"apple", "bananas", "cherry"}))
```

```python
print({'apple', 'banana', 'orange', 'pear'} | {'grape', 'banana', 'peach', 'orange'})
```

```python
print({"a", "b", "c"} | {1, 2, 3})
```

```python
print({"a", "b", "c"} | {1, 2, 3} | {"apple", "bananas", "cherry"})
```
