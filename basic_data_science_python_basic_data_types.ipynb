## Operator Precedence

The table below shows the precedence and associativity of arithmetic operators

| Operator | Description | Associativity |
| - | - | - |
| `**` | exponentiation | right to left |
| `%`, `*`, `/`, `//` | remainder, multiplication, division and integer division | left to right |
| `+`, `-` | addition and subtraction | left to right |
| `<`, `<=`, `>`, `>=`, `==`, `!=` , `in`, `not in` | comparisons | left to right |
| `not` | logical operator | left to right |
| `and` | logical operator | left to right |
| `or` | logical operator | left to right |

## `None` type
`None` is a special value that indicates the absence of a meaningful value.
It is used the way `True` and `False` are used for booleans.
If `None` is used in an expression, it causes an error.
It's only use is to test if a variable has no value.

There are two ways to test if a value is `None`:

| Method | Description |
| - | - |
| *variable* `is None` | Whether a value is `None` |
| *variable* `is not None` | Whether a value is not `None` |

#### *variable* `is None`
Whether a variable has the value `None`.

```python
not_none = 3
print(not_none is None)
```

#### *variable* `is not None`
Whether a variable does not have the value `None`.

```python
not_none = 3
print(not_none is not None)
```
