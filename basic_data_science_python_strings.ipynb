## Escape Sequences
A backslash `\` in a string "escapes" the normal way it reads and interprets the next character, and gives it a new meaning as a special character or to include formatting in a string.

These are the most common backslash escapes:

| Escape sequence | Name | Use |
| - | - | - |
| \n | newline (line break) | break text across multiple lines |
| \t | tab (indent) | indent text to a fixed next tab stop |
| \\' | single quote | Insert a single quote in a single quoted string |
| \” | double quote | Insert a double quote in a double quoted string |
| \\\ | backslash |  Insert a backslash in a single or double quoted string |

### Raw Strings
Raw strings ignore backslashes in escape sequences, allowing them to be included as-is in the string. They begin with `r` followed by a single or double quoted string. Between the starting and ending quotes, backslashes `\` are not treated as special.
- `r'...'`
- `r"..."`

````python
print(r'single quoted raw strings ignore backslashes \ in escape sequences like \n and \t')
print(r"double quoted raw strings ignore backslashes \ in escape sequences like \n and \t")
````

Raw strings beginning with `r` followed by a triple quotes (single or double quotes) can include single and double quotes without escaping them.
- `r'''...'''`
- `r"""..."""`

````python
print(r'''triple single quoted raw strings ignore escape sequences and can include single quotes ' ''')
print(r"""triple double quoted raw strings ignore escape sequences and can include double quotes " """)
````

### Indexing Strings
Individual characters can be accessed using a single position inside square brackets `[]`.

````python
# the first three vowels
print("DON'T PANIC!"[1])
print("DON'T PANIC!"[7])
print("DON'T PANIC!"[9])

# the first three vowels with negative indexes
print("DON'T PANIC!"[-11])
print("DON'T PANIC!"[-5])
print("DON'T PANIC!"[-3])
````

### String Slices
String *slices* access sequences of characters between positions in square brackets separated by one or more colons `:`.
The ways to use slices are:
- *string*`[`*start*`:`*end* `]`
- *string*`[`*start*`:`*end* `:`*step*`]`

The characters are selected from the start position up to, but not including, the end position.
If a step is specified, that many characters are skipped for each character selected between the start and end.

````python
# the first word
print("DON'T PANIC!"[0:6])

# the second word
print("DON'T PANIC!"[6:12])
````

- __Exercise__: Print out the words using negative positions.

````python
# every other characters of the first word
print("DON'T PANIC!"[0:6:2])

# every other characters of the second word
print("DON'T PANIC!"[6:12:2])
````

- __Exercise__: Print out every other character of the words using negative positions.

Either the first or second position may be omitted:
- If the first position is omitted, the slice starts at the beginning of the string, up to but not including the second position.
- If the second position is omitted, the slice starts at the first character and ends with the last character of the string.

````python
# the first word with the first position omitted
print("DON'T PANIC!"[:5])

# the second word with the second position omitted
print("DON'T PANIC!"[7:])
````

## Searching Strings
There are two functions to search strings for a substring:

| Function | Description |
| - | - |
| `find()` | Find the starting position of the first occurrence of a substring |
| `rfind()` | Find the starting position of the last occurrence of a substring |


### `find()`
There are three ways to use `find()`:

| use | meaning |
| - | - |
| `find(`*substring*`)` | Find first position of *substring* |
| `find(`*substring*`, `*start*`)` |  Find first position of *substring* beginning at *start* |
| `find(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* between *start* and *end* |

`find()` returns position -1 if *substring* is not found.

````python
# find the position of first PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC"))

# find the position of second PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 14))

# find the position of PANIC between positions 14 and 26
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 14, 26))

# the position of PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".find("PANIC", 35))
````

__Exercise__: find the word using negative positions.

### `rfind()`
There are three ways to use `rfind()`:

| use | meaning |
| - | - |
| `rfind(`*substring*`)` | Find first position of *substring* from the end|
| `rfind(`*substring*`, `*start*`)` |  Find first position of *substring* from the end beginning at *start* |
| `rfind(`*substring*`, `*start*`, `*end*`)` |  Find first position of *substring* from the end between *start* and *end* |

`rfind()` returns position -1 if *substring* is not found.

````python
# find the position of rightmost PANIC
#      0         1         2         3
#      01234567890123456789012345678901234567
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC"))
# find the position of rightmost starting at position 14 PANIC
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 14))
# find the position of rightmost PANIC between positions 14 and 26
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 14, 26))
# the position of rightmost PANIC if not found
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".rfind("PANIC", 35))
````

## Trimming Strings
There are three functions to trim strings:

| Function | Description |
| - | - |
| `lstrip()` | The string with leading characters removed |
| `rstrip()` | The string with trailing characters removed |
| `strip()` | The string with leading and trailing characters removed |


### `lstrip()`
There are two ways to use `lstrip()`:

| use | meaning |
| - | - |
| `lstrip()` | Remove leading whitespace (space and tabs) |
| `lstrip(`*characters*`)` |  Remove any of the given leading *characters* |

````python
# remove leading spaces
print("   DON'T PANIC!".lstrip())

# remove leading "   www."
print("   www.google.com".lstrip(" w."))
````

### `rstrip()`
There are two ways to use `rstrip()`:

| use | meaning |
| - | - |
| `rstrip()` | Remove trailing whitespace (space and tabs) |
| `rstrip(`*characters*`)` |  Remove any of the given trailing *characters* |

````python
# remove trailing spaces
print("DON'T PANIC!   ".rstrip())

# remove trailing ".com   "
print("www.google.com   ".rstrip(" .com"))
````

### `strip()`
There are two ways to use `strip()`:

| use | meaning |
| - | - |
| `strip()` | Remove trailing whitespace (space and tabs) |
| `strip(`*characters*`)` |  Remove any of the given trailing *characters* |

````python
# remove leading and trailing spaces
print("   DON'T PANIC!   ".strip())

# remove leading and trailing characters
print("   www.google.com   ".strip(' cmowz.'))
````

## Changing Case
There are five functions to change the case of string characters:

| Function | Description |
| - | - |
| `upper()` | Convert characters to UPPER CASE |
| `lower()` | Convert characters to lower case |
| `capitalize()` | Capitalize just the first word |
| `title()` | Capitalize each word |
| `swapcase()` | Invert the case of each word  |


### `upper()`
Convert characters to UPPER CASE.

````python
# converted all words to uppercase
print("So Long, and Thanks for All the Fish".upper())
````

### `lower()`
Convert characters to lower case.

````python
# converted all words to lowercase
print("So Long, and Thanks for All the Fish".lower())
````

### `capitalize()`
Capitalize just the first word.

````python
# capitalize just the first word
print("So Long, and Thanks for All the Fish".capitalize())
````

### `title()`
Capitalize each word.

````python
# capitalize each word
print("so long, and thanks for all the fish".title())
````

### `swapcase()`
Flip the case of each word.

````python
# invert case of each word
print("So Long, and Thanks for All the Fish".swapcase())
````

## String Properties
There are two functions to evaluate string properties:

| Function | Description |
| - | - |
| `count()` | The number of times a substring occurs in a string |
| `len()` | The number of characters in a string |


### `count()`
The number of times a substring occurs in a string.

````python
# count the instances of a substring
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!"))

# count the instances of a substring starting at a character
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!", 4))

# count the instances of a substring with starting and ending positions
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".count("PANIC!", 9, 14))
````

### `len()`
The number of characters in a string.

````python
# the length of a string
len("DON'T PANIC!")
````

## Formatting Strings
There are four functions to format strings:

| Function | Description |
| - | - |
| `center()` | The string centered in a string of a given width |
| `ljust()` | The string flush left in a string of a given width |
| `rjust()` | The string flush right in a string of a given width |
| `zfill()` | The numeric string left filled with zeros in a string of a given width |


### `center()`
There are two ways to use `center()`:

| use | meaning |
| - | - |
| `center(*width*)` | The string centered in a blank string of a given width |
| `center(`*width*`, `*fillchar*`)` |  The string centered in a string of a given width filled out with *fillchar* |

````python
print("DON'T PANIC!".center(20))
print("DON'T PANIC!".center(20, '-'))
````

### `ljust()`
There are two ways to use `ljust()`:

| use | meaning |
| - | - |
| `ljust(*width*)` | The string flush left in a blank string of a given width |
| `ljust(`*width*`, `*fillchar*`)` |  The string flush left in a string of a given width filled out with *fillchar* |

````python
print("DON'T PANIC!".ljust(20))
print("DON'T PANIC!".ljust(20, '-'))
````

### `rjust()`
There are two ways to use `rjust()`:

| use | meaning |
| - | - |
| `rjust(*width*)` | The string flush right in a blank string of a given width |
| `rjust(`*width*`, `*fillchar*`)` |  The string right left in a string of a given width filled out with *fillchar* |

````python
print("DON'T PANIC!".rjust(20))
print("DON'T PANIC!".rjust(20, '-'))
````

### `zfill(`*width*`)`
The string left filled with zeros in a string of a given width.
If the string begins with a minus sign, the zeros are placed after it.


````python
print("42".zfill(6))
print("-42".zfill(6))
````

## Modifying Strings
These are four functions to modify strings:

| Function | Description |
| - | - |
| `removeprefix()` | The string with a leading substring removed |
| `removesuffix()` | The string with a trailing substring removed |
| `replace()` | The string with a substring replaced a given of times |
| `join()` | The string with characters separated with a given substring |

### `removeprefix(`*substring*`)`
The string with a leading substring removed.

````python
# strings can have a fixed leading substring removed
print("So Long, and Thanks for All the Fish".removeprefix('So Long, and '))
````

### `removesuffix(`*substring*`)`
The string with a trailing substring removed.

````python
# strings can have a fixed ending substring removed
print("So Long, and Thanks for All the Fish".removesuffix(', and Thanks for All the Fish'))
````

### `replace()`
There are two ways to use `replace()`:

| use | meaning |
| - | - |
| `replace(`*old*`, `*new*`)` | The string with all substrings *old* replaced by *new* |
| `replace(`*old*`, `*new*`, `*count*`)` |  The string a given number of substrings *old* replaced by *new* |

````python
# all string substrings can be replaced
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".replace('PANIC', 'BE SCARED'))

# a fixed number of string substrings can be replaced
print("DON'T PANIC! DON'T PANIC! DON'T PANIC!".replace('PANIC', 'SCARED', 2))
````

### `join(`*substring*`)`
The string with characters separated with a given substring.

````python
# join string characters separated by a substring
print("-".join("DON'T PANIC!"))
````
